# Analytic Geometry: The Bridge Between Algebra and Geometry

## Why Analytic Geometry Matters for Calculus

Analytic geometry connects algebraic equations with geometric shapes, which is **essential** for calculus:

- **Multivariable Calculus**: Understanding 3D space, planes, and surfaces
- **Optimization**: Finding maximum/minimum values on curves and surfaces
- **Parametric Curves**: Describing motion and computing arc length using integration
- **Vector Calculus**: Gradient, divergence, curl - all rooted in analytic geometry
- **Conic Sections**: Used in physics (orbits, optics) and optimization

**The Big Idea**: Every geometric object can be described algebraically, and every equation represents a geometric shape. This duality lets us use calculus tools on geometric problems!

## 1. Coordinate Systems

### Cartesian Coordinates (Rectangular)

The standard $(x, y)$ system:
- Point location: distance from perpendicular axes
- **2D**: $(x, y)$ 
- **3D**: $(x, y, z)$

### Polar Coordinates

Location by angle and distance: $(r, \theta)$
- $r$ = distance from origin
- $\theta$ = angle from positive x-axis

### Conversion Formulas

**Polar to Cartesian**:
$$x = r\cos\theta$$
$$y = r\sin\theta$$

**Cartesian to Polar**:
$$r = \sqrt{x^2 + y^2}$$
$$\theta = \arctan\left(\frac{y}{x}\right) \quad \text{(with quadrant adjustments)}$$

**Why polar?** Many curves (circles, spirals, roses) are simpler in polar coordinates!

```python
# Visualize coordinate systems
fig = plt.figure(figsize=(18, 6))

# Plot 1: Cartesian coordinates
ax1 = fig.add_subplot(131)
point_x, point_y = 3, 2
ax1.plot(point_x, point_y, 'ro', markersize=15, label=f'Point ({point_x}, {point_y})')
ax1.plot([0, point_x], [0, 0], 'b--', linewidth=2, alpha=0.6)
ax1.plot([point_x, point_x], [0, point_y], 'g--', linewidth=2, alpha=0.6)
ax1.text(point_x/2, -0.3, f'x = {point_x}', fontsize=11, ha='center', color='blue')
ax1.text(point_x + 0.3, point_y/2, f'y = {point_y}', fontsize=11, color='green')
ax1.axhline(y=0, color='k', linewidth=1)
ax1.axvline(x=0, color='k', linewidth=1)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(-1, 5)
ax1.set_ylim(-1, 4)
ax1.set_xlabel('x-axis', fontsize=12)
ax1.set_ylabel('y-axis', fontsize=12)
ax1.set_title('Cartesian Coordinates', fontsize=14, fontweight='bold')
ax1.legend()
ax1.set_aspect('equal')

# Plot 2: Polar coordinates  
ax2 = fig.add_subplot(132, projection='polar')
r_val = np.sqrt(point_x**2 + point_y**2)
theta_val = np.arctan2(point_y, point_x)
ax2.plot(theta_val, r_val, 'ro', markersize=15, label=f'Point (r={r_val:.2f}, θ={np.degrees(theta_val):.1f}°)')
ax2.plot([0, theta_val], [0, r_val], 'r-', linewidth=3)
# Draw angle arc
arc_theta = np.linspace(0, theta_val, 50)
ax2.plot(arc_theta, np.ones_like(arc_theta) * 0.8, 'g-', linewidth=2)
ax2.set_title('Polar Coordinates\nSame Point!', fontsize=14, fontweight='bold')
ax2.legend(loc='upper right')
ax2.grid(True)

# Plot 3: Both overlaid
ax3 = fig.add_subplot(133)
ax3.plot(point_x, point_y, 'ro', markersize=15, label=f'Point', zorder=5)
# Cartesian
ax3.plot([0, point_x], [0, 0], 'b-', linewidth=2, alpha=0.7, label='Cartesian: x, y')
ax3.plot([point_x, point_x], [0, point_y], 'b-', linewidth=2, alpha=0.7)
# Polar
ax3.plot([0, point_x], [0, point_y], 'r-', linewidth=3, alpha=0.7, label='Polar: r')
arc_angles = np.linspace(0, theta_val, 50)
ax3.plot(0.8 * np.cos(arc_angles), 0.8 * np.sin(arc_angles), 'g-', linewidth=2, label='Polar: θ')
ax3.axhline(y=0, color='k', linewidth=1)
ax3.axvline(x=0, color='k', linewidth=1)
ax3.grid(True, alpha=0.3)
ax3.set_xlim(-1, 5)
ax3.set_ylim(-1, 4)
ax3.set_xlabel('x', fontsize=12)
ax3.set_ylabel('y', fontsize=12)
ax3.set_title('Both Systems Overlay', fontsize=14, fontweight='bold')
ax3.legend()
ax3.set_aspect('equal')

plt.tight_layout()
plt.show()

print(f"Conversion Example:")
print(f"  Cartesian: ({point_x}, {point_y})")
print(f"  Polar: (r = {r_val:.3f}, θ = {np.degrees(theta_val):.2f}° = {theta_val:.3f} rad)")
print(f"\nVerify: x = r·cos(θ) = {r_val:.3f}·cos({theta_val:.3f}) = {r_val * np.cos(theta_val):.3f}")
print(f"        y = r·sin(θ) = {r_val:.3f}·sin({theta_val:.3f}) = {r_val * np.sin(theta_val):.3f}")
```

### 3D Cartesian Coordinates

Extend to three dimensions: $(x, y, z)$

- **Distance formula**: $d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2 + (z_2-z_1)^2}$
- **Midpoint**: $M = \left(\frac{x_1+x_2}{2}, \frac{y_1+y_2}{2}, \frac{z_1+z_2}{2}\right)$

Other 3D systems:
- **Cylindrical**: $(r, \theta, z)$ - polar in x-y plane, height z
- **Spherical**: $(\rho, \theta, \phi)$ - distance and two angles

```python
# 3D coordinate visualization
fig = plt.figure(figsize=(14, 6))

# 3D point
ax1 = fig.add_subplot(121, projection='3d')
point_3d = np.array([3, 2, 4])
ax1.scatter(*point_3d, color='red', s=200, label=f'Point {tuple(point_3d)}')

# Draw lines to axes
ax1.plot([0, point_3d[0]], [0, 0], [0, 0], 'b--', linewidth=2, alpha=0.6)
ax1.plot([point_3d[0], point_3d[0]], [0, point_3d[1]], [0, 0], 'g--', linewidth=2, alpha=0.6)
ax1.plot([point_3d[0], point_3d[0]], [point_3d[1], point_3d[1]], [0, point_3d[2]], 'r--', linewidth=2, alpha=0.6)

# Draw box outline
vertices = [[0,0,0], [point_3d[0],0,0], [point_3d[0],point_3d[1],0], 
            [0,point_3d[1],0], [0,0,point_3d[2]], [point_3d[0],0,point_3d[2]], 
            list(point_3d), [0,point_3d[1],point_3d[2]]]
edges = [[0,1],[1,2],[2,3],[3,0],[4,5],[5,6],[6,7],[7,4],[0,4],[1,5],[2,6],[3,7]]
for edge in edges:
    pts = np.array([vertices[edge[0]], vertices[edge[1]]])
    ax1.plot3D(pts[:,0], pts[:,1], pts[:,2], 'gray', linewidth=1, alpha=0.3)

ax1.set_xlabel('X', fontsize=12, fontweight='bold')
ax1.set_ylabel('Y', fontsize=12, fontweight='bold')
ax1.set_zlabel('Z', fontsize=12, fontweight='bold')
ax1.set_title('3D Cartesian Coordinates', fontsize=14, fontweight='bold')
ax1.legend()

# Distance calculation
ax2 = fig.add_subplot(122, projection='3d')
p1 = np.array([1, 1, 1])
p2 = np.array([4, 5, 3])
ax2.scatter(*p1, color='blue', s=150, label=f'P1 = {tuple(p1)}')
ax2.scatter(*p2, color='red', s=150, label=f'P2 = {tuple(p2)}')
ax2.plot([p1[0], p2[0]], [p1[1], p2[1]], [p1[2], p2[2]], 'g-', linewidth=3, 
         label=f'Distance = {np.linalg.norm(p2-p1):.2f}')

# Midpoint
midpoint = (p1 + p2) / 2
ax2.scatter(*midpoint, color='purple', s=150, marker='s', label=f'Midpoint = ({midpoint[0]:.1f}, {midpoint[1]:.1f}, {midpoint[2]:.1f})')

ax2.set_xlabel('X', fontsize=12)
ax2.set_ylabel('Y', fontsize=12)
ax2.set_zlabel('Z', fontsize=12)
ax2.set_title('Distance and Midpoint in 3D', fontsize=14, fontweight='bold')
ax2.legend()

plt.tight_layout()
plt.show()

dist = np.linalg.norm(p2 - p1)
print(f"Distance formula: d = √[(x₂-x₁)² + (y₂-y₁)² + (z₂-z₁)²]")
print(f"  = √[({p2[0]}-{p1[0]})² + ({p2[1]}-{p1[1]})² + ({p2[2]}-{p1[2]})²]")
print(f"  = √[{(p2[0]-p1[0])**2} + {(p2[1]-p1[1])**2} + {(p2[2]-p1[2])**2}] = √{sum((p2-p1)**2):.0f} = {dist:.3f}")
```

## 2. Lines in the Plane

### Forms of a Line Equation

**Slope-Intercept Form**:
$$y = mx + b$$
- $m$ = slope
- $b$ = y-intercept

**Point-Slope Form**:
$$y - y_1 = m(x - x_1)$$
Useful when you know a point and the slope

**General Form**:
$$Ax + By + C = 0$$
Most general representation

**Vector Form**:
$$\vec{r}(t) = \vec{r}_0 + t\vec{d}$$
where $\vec{r}_0$ is a point on the line and $\vec{d}$ is the direction vector

### Parallel and Perpendicular Lines

- **Parallel**: Same slope ($m_1 = m_2$)
- **Perpendicular**: $m_1 \cdot m_2 = -1$ (negative reciprocals)

```python
# Visualize different line forms and relationships
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

x = np.linspace(-2, 8, 1000)

# Plot 1: Different forms of the same line
m, b = 2, -3
y = m * x + b
axes[0, 0].plot(x, y, 'b-', linewidth=3, label=f'y = {m}x + ({b})')
axes[0, 0].plot(0, b, 'ro', markersize=12, label=f'y-intercept = {b}')
axes[0, 0].plot(2, m*2+b, 'go', markersize=10)
axes[0, 0].plot([2, 3], [m*2+b, m*2+b], 'g--', linewidth=2, label='run = 1')
axes[0, 0].plot([3, 3], [m*2+b, m*3+b], 'r--', linewidth=2, label=f'rise = {m}')
axes[0, 0].axhline(y=0, color='k', linewidth=0.5)
axes[0, 0].axvline(x=0, color='k', linewidth=0.5)
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_xlim(-2, 6)
axes[0, 0].set_ylim(-8, 8)
axes[0, 0].set_title(f'Slope-Intercept Form\ny = mx + b (m={m}, b={b})', fontsize=13, fontweight='bold')
axes[0, 0].legend()

# Plot 2: Point-slope form
x0, y0 = 3, 4
m2 = -1.5
y2 = y0 + m2 * (x - x0)
axes[0, 1].plot(x, y2, 'purple', linewidth=3, label=f'y - {y0} = {m2}(x - {x0})')
axes[0, 1].plot(x0, y0, 'ro', markersize=15, label=f'Point ({x0}, {y0})')
axes[0, 1].axhline(y=0, color='k', linewidth=0.5)
axes[0, 1].axvline(x=0, color='k', linewidth=0.5)
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_xlim(-1, 7)
axes[0, 1].set_ylim(-4, 10)
axes[0, 1].set_title('Point-Slope Form\ny - y₁ = m(x - x₁)', fontsize=13, fontweight='bold')
axes[0, 1].legend()

# Plot 3: Parallel lines
m_par = 1.5
b_par = [0, 2, -2]
colors = ['blue', 'green', 'red']
for b_val, color in zip(b_par, colors):
    y_par = m_par * x + b_val
    axes[1, 0].plot(x, y_par, color=color, linewidth=2.5, label=f'y = {m_par}x + {b_val}')
axes[1, 0].axhline(y=0, color='k', linewidth=0.5)
axes[1, 0].axvline(x=0, color='k', linewidth=0.5)
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_xlim(-2, 6)
axes[1, 0].set_ylim(-4, 12)
axes[1, 0].set_title(f'Parallel Lines\nSame slope: m = {m_par}', fontsize=13, fontweight='bold')
axes[1, 0].legend()

# Plot 4: Perpendicular lines
m_perp1 = 2
m_perp2 = -1/m_perp1  # Negative reciprocal
y_perp1 = m_perp1 * x + 1
y_perp2 = m_perp2 * x + 3
axes[1, 1].plot(x, y_perp1, 'b-', linewidth=3, label=f'y = {m_perp1}x + 1')
axes[1, 1].plot(x, y_perp2, 'r-', linewidth=3, label=f'y = {m_perp2}x + 3')
# Mark intersection
x_int = (3 - 1) / (m_perp1 - m_perp2)
y_int = m_perp1 * x_int + 1
axes[1, 1].plot(x_int, y_int, 'go', markersize=15, label='Intersection (90°)')
axes[1, 1].axhline(y=0, color='k', linewidth=0.5)
axes[1, 1].axvline(x=0, color='k', linewidth=0.5)
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].set_xlim(-2, 6)
axes[1, 1].set_ylim(-2, 10)
axes[1, 1].set_title(f'Perpendicular Lines\nm₁·m₂ = {m_perp1}·{m_perp2} = -1', fontsize=13, fontweight='bold')
axes[1, 1].legend()

plt.tight_layout()
plt.show()

print("Line Equations Summary:")
print(f"  Slope-intercept: y = mx + b")
print(f"  Point-slope:     y - y₁ = m(x - x₁)")
print(f"  General:         Ax + By + C = 0")
print(f"\nParallel lines: m₁ = m₂")
print(f"Perpendicular:  m₁ · m₂ = -1")
```

## 3. Planes in 3D Space

### Equation of a Plane

**General Form**:
$$Ax + By + Cz + D = 0$$

**Normal Vector Form**:
$$\vec{n} \cdot (\vec{r} - \vec{r}_0) = 0$$

where:
- $\vec{n} = \langle A, B, C \rangle$ is the **normal vector** (perpendicular to the plane)
- $\vec{r}_0$ is any point on the plane

**Key Insight**: The coefficients $(A, B, C)$ give the direction perpendicular to the plane!

```python
# Visualize planes in 3D
fig = plt.figure(figsize=(16, 6))

# Create meshgrid for planes
x_range = np.linspace(-3, 3, 50)
y_range = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x_range, y_range)

# Plot 1: Single plane with normal vector
ax1 = fig.add_subplot(131, projection='3d')
# Plane: 2x + 3y + z = 6
Z1 = 6 - 2*X - 3*Y
ax1.plot_surface(X, Y, Z1, alpha=0.5, cmap='viridis')

# Normal vector
normal = np.array([2, 3, 1])
point_on_plane = np.array([0, 0, 6])
ax1.quiver(*point_on_plane, *normal, length=2, color='red', arrow_length_ratio=0.3, linewidth=3,
          label='Normal vector (2,3,1)')
ax1.scatter(*point_on_plane, color='blue', s=100)

ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')
ax1.set_title('Plane: 2x + 3y + z = 6\nNormal: (2, 3, 1)', fontsize=12, fontweight='bold')
ax1.legend()

# Plot 2: Intersection of two planes
ax2 = fig.add_subplot(132, projection='3d')
# Plane 1: z = 2
Z2a = 2 * np.ones_like(X)
# Plane 2: x + y + z = 4
Z2b = 4 - X - Y
ax2.plot_surface(X, Y, Z2a, alpha=0.4, color='blue', label='z = 2')
ax2.plot_surface(X, Y, Z2b, alpha=0.4, color='red', label='x + y + z = 4')

# Intersection line: z=2 and x+y=2
t_line = np.linspace(-3, 3, 100)
x_line = t_line
y_line = 2 - t_line
z_line = 2 * np.ones_like(t_line)
ax2.plot(x_line, y_line, z_line, 'g-', linewidth=4, label='Intersection line')

ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Z')
ax2.set_title('Two Planes Intersecting\nin a Line', fontsize=12, fontweight='bold')

# Plot 3: Three coordinate planes
ax3 = fig.add_subplot(133, projection='3d')
# xy-plane (z=0)
Z3a = np.zeros_like(X)
ax3.plot_surface(X, Y, Z3a, alpha=0.3, color='red')
ax3.text(0, 0, 0, 'xy-plane\n(z=0)', fontsize=10, ha='center')

# xz-plane (y=0)
Y3b = np.zeros_like(X)
x_temp, z_temp = np.meshgrid(x_range, x_range)
ax3.plot_surface(x_temp, Y3b[:len(x_range),:len(x_range)], z_temp, alpha=0.3, color='green')
ax3.text(0, 0, 0, '   xz-plane\n   (y=0)', fontsize=10, ha='left')

# yz-plane (x=0)
X3c = np.zeros_like(Y)
y_temp, z_temp = np.meshgrid(y_range, y_range)
ax3.plot_surface(X3c[:len(y_range),:len(y_range)], y_temp, z_temp, alpha=0.3, color='blue')
ax3.text(0, 0, 0, '      yz-plane\n      (x=0)', fontsize=10, ha='left')

ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_zlabel('Z')
ax3.set_title('Three Coordinate Planes', fontsize=12, fontweight='bold')
ax3.set_xlim([-3, 3])
ax3.set_ylim([-3, 3])
ax3.set_zlim([-3, 3])

plt.tight_layout()
plt.show()

print("Key Points:")
print("  • A plane in 3D requires 3 numbers (A, B, C) to describe its orientation")
print("  • The normal vector (A, B, C) is perpendicular to every line in the plane")
print("  • Two planes intersect in a line (unless parallel)")
print("  • Three planes can intersect at a point, line, or not at all")
```

## 4. Conic Sections

Conic sections are curves obtained by intersecting a plane with a double cone. They appear everywhere in nature and physics!

### The Four Conics:

1. **Circle**: All points equidistant from a center
2. **Ellipse**: Sum of distances to two foci is constant
3. **Parabola**: Distance to focus equals distance to directrix
4. **Hyperbola**: Difference of distances to two foci is constant

### Why They Matter:
- **Planetary orbits**: Ellipses (Kepler's laws)
- **Projectile motion**: Parabolas
- **Satellite dishes**: Parabolas (focus property)
- **Navigation**: Hyperbolas (LORAN)
- **Optimization**: Calculus on conics

### Circle

**Standard Form**:
$$(x - h)^2 + (y - k)^2 = r^2$$

- Center: $(h, k)$
- Radius: $r$

**General Form**:
$$x^2 + y^2 + Dx + Ey + F = 0$$

**Calculus Application**: Find tangent line at a point using implicit differentiation!

```python
# Circle visualization with calculus
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Circle with key features
h, k, r = 2, 1, 3
theta = np.linspace(0, 2*np.pi, 1000)
x_circle = h + r * np.cos(theta)
y_circle = k + r * np.sin(theta)

axes[0].plot(x_circle, y_circle, 'b-', linewidth=3, label=f'(x-{h})² + (y-{k})² = {r}²')
axes[0].plot(h, k, 'ro', markersize=15, label=f'Center ({h}, {k})')
axes[0].plot([h, h+r], [k, k], 'r--', linewidth=2, label=f'Radius = {r}')
axes[0].axhline(y=0, color='k', linewidth=0.5)
axes[0].axvline(x=0, color='k', linewidth=0.5)
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(-2, 7)
axes[0].set_ylim(-3, 5)
axes[0].set_aspect('equal')
axes[0].set_title('Circle: Standard Form', fontsize=14, fontweight='bold')
axes[0].legend()

# Plot 2: Tangent line using calculus
axes[1].plot(x_circle, y_circle, 'b-', linewidth=3, label='Circle')
axes[1].plot(h, k, 'ro', markersize=10)

# Point on circle
angle_point = np.pi/4
x_p = h + r * np.cos(angle_point)
y_p = k + r * np.sin(angle_point)
axes[1].plot(x_p, y_p, 'go', markersize=15, label=f'Point ({x_p:.2f}, {y_p:.2f})')

# Tangent line: implicit differentiation of (x-h)² + (y-k)² = r²
# gives: 2(x-h) + 2(y-k)dy/dx = 0
# so: dy/dx = -(x-h)/(y-k)
slope_tangent = -(x_p - h) / (y_p - k)
x_tangent = np.linspace(x_p - 2, x_p + 2, 100)
y_tangent = y_p + slope_tangent * (x_tangent - x_p)
axes[1].plot(x_tangent, y_tangent, 'r-', linewidth=2, label=f'Tangent (slope={slope_tangent:.2f})')

# Radius to point (perpendicular to tangent)
axes[1].plot([h, x_p], [k, y_p], 'purple', linewidth=2, linestyle='--', label='Radius ⊥ Tangent')

axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].axvline(x=0, color='k', linewidth=0.5)
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(-2, 7)
axes[1].set_ylim(-3, 5)
axes[1].set_aspect('equal')
axes[1].set_title('Tangent Line via Calculus\n(Implicit Differentiation)', fontsize=14, fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()

print("Calculus on Circles:")
print(f"  Equation: (x-{h})² + (y-{k})² = {r}²")
print(f"  Implicit differentiation: 2(x-{h}) + 2(y-{k})dy/dx = 0")
print(f"  Slope at ({x_p:.2f}, {y_p:.2f}): dy/dx = -({x_p:.2f}-{h})/({y_p:.2f}-{k}) = {slope_tangent:.2f}")
print(f"\n  Note: Radius has slope {(y_p-k)/(x_p-h):.2f}")
print(f"        Product of slopes: {slope_tangent:.2f} × {(y_p-k)/(x_p-h):.2f} = {slope_tangent * (y_p-k)/(x_p-h):.2f} ≈ -1 ✓")
```

### Ellipse

**Standard Form** (center at origin):
$$\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1$$

- $a$ = semi-major axis (larger)
- $b$ = semi-minor axis (smaller)
- Foci at $(\pm c, 0)$ where $c^2 = a^2 - b^2$ (if $a > b$)

**Key Property**: Sum of distances to foci is constant = $2a$

**Applications**: Planetary orbits, whispering galleries, lithotripsy

```python
# Ellipse visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Ellipse with foci
a, b = 4, 2.5
c = np.sqrt(a**2 - b**2)

theta_ell = np.linspace(0, 2*np.pi, 1000)
x_ell = a * np.cos(theta_ell)
y_ell = b * np.sin(theta_ell)

axes[0].plot(x_ell, y_ell, 'b-', linewidth=3, label=f'x²/{a}² + y²/{b:.1f}² = 1')
axes[0].plot([-c, c], [0, 0], 'ro', markersize=15, label=f'Foci (±{c:.2f}, 0)')

# Show constant sum property
angle_demo = np.pi / 3
x_demo = a * np.cos(angle_demo)
y_demo = b * np.sin(angle_demo)
axes[0].plot(x_demo, y_demo, 'go', markersize=12)
d1 = np.sqrt((x_demo + c)**2 + y_demo**2)
d2 = np.sqrt((x_demo - c)**2 + y_demo**2)
axes[0].plot([x_demo, -c], [y_demo, 0], 'g--', linewidth=2, label=f'd₁ = {d1:.2f}')
axes[0].plot([x_demo, c], [y_demo, 0], 'purple', linestyle='--', linewidth=2, label=f'd₂ = {d2:.2f}')

# Draw axes
axes[0].plot([-a, a], [0, 0], 'k-', linewidth=1.5, alpha=0.5)
axes[0].plot([0, 0], [-b, b], 'k-', linewidth=1.5, alpha=0.5)
axes[0].text(a, -0.3, f'a = {a}', fontsize=11, ha='center')
axes[0].text(0.2, b, f'b = {b}', fontsize=11)

axes[0].axhline(y=0, color='k', linewidth=0.5)
axes[0].axvline(x=0, color='k', linewidth=0.5)
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(-5, 5)
axes[0].set_ylim(-4, 4)
axes[0].set_aspect('equal')
axes[0].set_title(f'Ellipse\nd₁ + d₂ = {d1+d2:.2f} = 2a = {2*a}', fontsize=14, fontweight='bold')
axes[0].legend()

# Plot 2: Ellipse area using integration
axes[1].fill(x_ell, y_ell, alpha=0.3, color='lightblue', label=f'Area = πab = π·{a}·{b:.1f} = {np.pi*a*b:.2f}')
axes[1].plot(x_ell, y_ell, 'b-', linewidth=3)

# Show how we'd integrate
x_int = 2
y_int = b * np.sqrt(1 - (x_int/a)**2)
axes[1].plot([x_int, x_int], [-y_int, y_int], 'r-', linewidth=3, label='Vertical slice for integration')
axes[1].plot(x_int, y_int, 'ro', markersize=10)
axes[1].plot(x_int, -y_int, 'ro', markersize=10)

axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].axvline(x=0, color='k', linewidth=0.5)
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(-5, 5)
axes[1].set_ylim(-4, 4)
axes[1].set_aspect('equal')
axes[1].set_title('Area via Integration\nA = ∫∫dA = πab', fontsize=14, fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()

print("Ellipse Properties:")
print(f"  Semi-major axis a = {a}")
print(f"  Semi-minor axis b = {b}")
print(f"  Focal distance c = √(a² - b²) = √({a}² - {b:.1f}²) = {c:.2f}")
print(f"  Eccentricity e = c/a = {c/a:.3f} (circle when e=0, more elongated as e→1)")
print(f"  Area = πab = {np.pi * a * b:.2f}")
```

### Parabola

**Standard Forms**:

Vertical axis:
$$(x - h)^2 = 4p(y - k)$$

Horizontal axis:
$$(y - k)^2 = 4p(x - h)$$

- Vertex: $(h, k)$
- Focus: distance $p$ from vertex along axis
- Directrix: line distance $p$ on other side

**Most common**: $y = ax^2 + bx + c$ (vertical parabola)

**Reflection Property**: Light from focus reflects parallel to axis - used in satellite dishes and headlights!

```python
# Parabola visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Parabola with focus and directrix
p = 1.5  # focal parameter
x_par = np.linspace(-4, 4, 1000)
y_par = x_par**2 / (4*p)

axes[0].plot(x_par, y_par, 'b-', linewidth=3, label=f'x² = 4py (p={p})')
axes[0].plot(0, 0, 'go', markersize=12, label='Vertex (0, 0)')
axes[0].plot(0, p, 'ro', markersize=15, label=f'Focus (0, {p})')
axes[0].axhline(y=-p, color='purple', linestyle='--', linewidth=2, label=f'Directrix y = -{p}')

# Show equal distance property
x_demo = 2
y_demo = x_demo**2 / (4*p)
axes[0].plot(x_demo, y_demo, 'orange', marker='o', markersize=12)
d_focus = np.sqrt(x_demo**2 + (y_demo - p)**2)
d_directrix = y_demo - (-p)
axes[0].plot([x_demo, 0], [y_demo, p], 'orange', linestyle='--', linewidth=2, 
            label=f'd to focus = {d_focus:.2f}')
axes[0].plot([x_demo, x_demo], [y_demo, -p], 'brown', linestyle='--', linewidth=2,
            label=f'd to directrix = {d_directrix:.2f}')

axes[0].axhline(y=0, color='k', linewidth=0.5)
axes[0].axvline(x=0, color='k', linewidth=0.5)
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(-5, 5)
axes[0].set_ylim(-3, 6)
axes[0].set_aspect('equal')
axes[0].set_title('Parabola Definition\nDistance to Focus = Distance to Directrix', fontsize=14, fontweight='bold')
axes[0].legend()

# Plot 2: Quadratic form with calculus
a_coef = 0.5
b_coef = -2
c_coef = 3
x_quad = np.linspace(-2, 6, 1000)
y_quad = a_coef * x_quad**2 + b_coef * x_quad + c_coef

# Vertex using calculus: dy/dx = 0
x_vertex = -b_coef / (2*a_coef)
y_vertex = a_coef * x_vertex**2 + b_coef * x_vertex + c_coef

axes[1].plot(x_quad, y_quad, 'b-', linewidth=3, label=f'y = {a_coef}x² + {b_coef}x + {c_coef}')
axes[1].plot(x_vertex, y_vertex, 'ro', markersize=15, label=f'Vertex ({x_vertex:.1f}, {y_vertex:.1f})')

# Tangent line at a point using derivative
x_tan = 4
y_tan = a_coef * x_tan**2 + b_coef * x_tan + c_coef
slope_tan = 2*a_coef*x_tan + b_coef
x_tan_line = np.linspace(x_tan - 2, x_tan + 2, 100)
y_tan_line = y_tan + slope_tan * (x_tan_line - x_tan)
axes[1].plot(x_tan_line, y_tan_line, 'g-', linewidth=2, label=f'Tangent at x={x_tan}\nslope = {slope_tan:.1f}')
axes[1].plot(x_tan, y_tan, 'go', markersize=12)

axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].axvline(x=0, color='k', linewidth=0.5)
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(-2, 6)
axes[1].set_ylim(-2, 8)
axes[1].set_title('Quadratic Form\nVertex: dy/dx = 0', fontsize=14, fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()

print("Finding Vertex with Calculus:")
print(f"  y = {a_coef}x² + {b_coef}x + {c_coef}")
print(f"  dy/dx = {2*a_coef}x + {b_coef}")
print(f"  Set dy/dx = 0: {2*a_coef}x + {b_coef} = 0")
print(f"  Vertex x-coordinate: x = {x_vertex:.1f}")
print(f"  Vertex y-coordinate: y = {y_vertex:.1f}")
```

### Hyperbola

**Standard Form** (center at origin, horizontal transverse axis):
$$\frac{x^2}{a^2} - \frac{y^2}{b^2} = 1$$

**Key Property**: |distance to focus₁ - distance to focus₂| = constant = $2a$

- Vertices: $(\pm a, 0)$
- Foci: $(\pm c, 0)$ where $c^2 = a^2 + b^2$
- Asymptotes: $y = \pm \frac{b}{a}x$

**Applications**: Navigation (LORAN), astronomy, cooling towers

```python
# Hyperbola visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Hyperbola with asymptotes
a_hyp = 2
b_hyp = 1.5
c_hyp = np.sqrt(a_hyp**2 + b_hyp**2)

# Right branch
x_right = np.linspace(a_hyp, 5, 500)
y_right_pos = b_hyp * np.sqrt(x_right**2/a_hyp**2 - 1)
y_right_neg = -y_right_pos

# Left branch
x_left = np.linspace(-5, -a_hyp, 500)
y_left_pos = b_hyp * np.sqrt(x_left**2/a_hyp**2 - 1)
y_left_neg = -y_left_pos

axes[0].plot(x_right, y_right_pos, 'b-', linewidth=3)
axes[0].plot(x_right, y_right_neg, 'b-', linewidth=3, label=f'x²/{a_hyp}² - y²/{b_hyp:.1f}² = 1')
axes[0].plot(x_left, y_left_pos, 'b-', linewidth=3)
axes[0].plot(x_left, y_left_neg, 'b-', linewidth=3)

# Asymptotes
x_asym = np.linspace(-5, 5, 100)
y_asym_pos = (b_hyp/a_hyp) * x_asym
y_asym_neg = -(b_hyp/a_hyp) * x_asym
axes[0].plot(x_asym, y_asym_pos, 'r--', linewidth=2, label=f'y = ±{b_hyp/a_hyp:.2f}x (asymptotes)')
axes[0].plot(x_asym, y_asym_neg, 'r--', linewidth=2)

# Foci and vertices
axes[0].plot([-c_hyp, c_hyp], [0, 0], 'ro', markersize=15, label=f'Foci (±{c_hyp:.2f}, 0)')
axes[0].plot([-a_hyp, a_hyp], [0, 0], 'go', markersize=12, label=f'Vertices (±{a_hyp}, 0)')

axes[0].axhline(y=0, color='k', linewidth=0.5)
axes[0].axvline(x=0, color='k', linewidth=0.5)
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(-5, 5)
axes[0].set_ylim(-4, 4)
axes[0].set_aspect('equal')
axes[0].set_title('Hyperbola with Asymptotes', fontsize=14, fontweight='bold')
axes[0].legend()

# Plot 2: Difference property
axes[1].plot(x_right, y_right_pos, 'b-', linewidth=3)
axes[1].plot(x_right, y_right_neg, 'b-', linewidth=3)
axes[1].plot(x_left, y_left_pos, 'b-', linewidth=3)
axes[1].plot(x_left, y_left_neg, 'b-', linewidth=3)
axes[1].plot([-c_hyp, c_hyp], [0, 0], 'ro', markersize=15)

# Demonstrate difference property
x_demo = 3
y_demo = b_hyp * np.sqrt(x_demo**2/a_hyp**2 - 1)
axes[1].plot(x_demo, y_demo, 'purple', marker='o', markersize=15)
d1_hyp = np.sqrt((x_demo + c_hyp)**2 + y_demo**2)
d2_hyp = np.sqrt((x_demo - c_hyp)**2 + y_demo**2)
axes[1].plot([x_demo, -c_hyp], [y_demo, 0], 'g--', linewidth=2, label=f'd₁ = {d1_hyp:.2f}')
axes[1].plot([x_demo, c_hyp], [y_demo, 0], 'orange', linestyle='--', linewidth=2, label=f'd₂ = {d2_hyp:.2f}')

axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].axvline(x=0, color='k', linewidth=0.5)
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(-5, 5)
axes[1].set_ylim(-4, 4)
axes[1].set_aspect('equal')
axes[1].set_title(f'Difference Property\n|d₁ - d₂| = {abs(d1_hyp - d2_hyp):.2f} = 2a = {2*a_hyp}', 
                 fontsize=14, fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()

print("Hyperbola Properties:")
print(f"  a = {a_hyp} (distance from center to vertex)")
print(f"  b = {b_hyp} (determines 'width' of opening)")
print(f"  c = √(a² + b²) = {c_hyp:.2f} (distance to foci)")
print(f"  Eccentricity e = c/a = {c_hyp/a_hyp:.3f} (always > 1 for hyperbola)")
print(f"  Asymptote slopes: ±b/a = ±{b_hyp/a_hyp:.2f}")
```

## 5. Parametric Equations

Instead of $y = f(x)$, express both coordinates as functions of a parameter $t$:
$$x = f(t), \quad y = g(t)$$

### Why Parametric?
- Describe curves that fail the vertical line test
- Natural for motion (t = time)
- Easy to compute arc length using calculus!

### Arc Length (Calculus!)
$$L = \int_a^b \sqrt{\left(\frac{dx}{dt}\right)^2 + \left(\frac{dy}{dt}\right)^2} \, dt$$

### Slope
$$\frac{dy}{dx} = \frac{dy/dt}{dx/dt}$$

```python
# Parametric curves
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: Circle as parametric
t_circle = np.linspace(0, 2*np.pi, 1000)
r_param = 2
x_param_circle = r_param * np.cos(t_circle)
y_param_circle = r_param * np.sin(t_circle)

axes[0, 0].plot(x_param_circle, y_param_circle, 'b-', linewidth=3, 
               label=f'x = {r_param}cos(t), y = {r_param}sin(t)')
# Show direction with arrows
for t_val in [0, np.pi/2, np.pi, 3*np.pi/2]:
    x_arrow = r_param * np.cos(t_val)
    y_arrow = r_param * np.sin(t_val)
    dx = -r_param * np.sin(t_val) * 0.3
    dy = r_param * np.cos(t_val) * 0.3
    axes[0, 0].arrow(x_arrow, y_arrow, dx, dy, head_width=0.2, head_length=0.15, fc='red', ec='red')

axes[0, 0].axhline(y=0, color='k', linewidth=0.5)
axes[0, 0].axvline(x=0, color='k', linewidth=0.5)
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_aspect('equal')
axes[0, 0].set_title('Circle as Parametric\n0 ≤ t ≤ 2π', fontsize=13, fontweight='bold')
axes[0, 0].legend()

# Plot 2: Cycloid (wheel rolling)
t_cycloid = np.linspace(0, 4*np.pi, 1000)
a_cycloid = 1
x_cycloid = a_cycloid * (t_cycloid - np.sin(t_cycloid))
y_cycloid = a_cycloid * (1 - np.cos(t_cycloid))

axes[0, 1].plot(x_cycloid, y_cycloid, 'purple', linewidth=3,
               label=f'x = a(t - sin t), y = a(1 - cos t)')
axes[0, 1].axhline(y=0, color='k', linewidth=0.5)
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_xlim(-0.5, 13)
axes[0, 1].set_ylim(-0.5, 2.5)
axes[0, 1].set_title('Cycloid (Rolling Wheel)\nFastest descent curve!', fontsize=13, fontweight='bold')
axes[0, 1].legend()

# Plot 3: Lissajous curve
t_liss = np.linspace(0, 2*np.pi, 1000)
A, B = 3, 2
a_liss, b_liss = 3, 2
delta = np.pi/2
x_liss = A * np.sin(a_liss * t_liss + delta)
y_liss = B * np.sin(b_liss * t_liss)

axes[1, 0].plot(x_liss, y_liss, 'green', linewidth=2.5,
               label=f'x = {A}sin({a_liss}t + π/2), y = {B}sin({b_liss}t)')
axes[1, 0].axhline(y=0, color='k', linewidth=0.5)
axes[1, 0].axvline(x=0, color='k', linewidth=0.5)
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_aspect('equal')
axes[1, 0].set_xlim(-4, 4)
axes[1, 0].set_ylim(-3, 3)
axes[1, 0].set_title('Lissajous Curve\nOscilloscope patterns', fontsize=13, fontweight='bold')
axes[1, 0].legend()

# Plot 4: Arc length calculation
t_arc = np.linspace(0, np.pi, 100)
x_arc = 2 * np.cos(t_arc)
y_arc = 2 * np.sin(t_arc)

axes[1, 1].plot(x_arc, y_arc, 'b-', linewidth=4, label='x = 2cos(t), y = 2sin(t)')

# Show derivative vectors at a point
t_pt = np.pi/4
x_pt = 2 * np.cos(t_pt)
y_pt = 2 * np.sin(t_pt)
dx_dt = -2 * np.sin(t_pt)
dy_dt = 2 * np.cos(t_pt)

axes[1, 1].plot(x_pt, y_pt, 'ro', markersize=15)
axes[1, 1].arrow(x_pt, y_pt, dx_dt*0.3, dy_dt*0.3, head_width=0.2, head_length=0.15,
                fc='red', ec='red', linewidth=2)
axes[1, 1].text(x_pt + 0.5, y_pt + 0.5, f'(dx/dt, dy/dt)\n= ({dx_dt:.2f}, {dy_dt:.2f})',
               fontsize=10, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

axes[1, 1].axhline(y=0, color='k', linewidth=0.5)
axes[1, 1].axvline(x=0, color='k', linewidth=0.5)
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].set_aspect('equal')
axes[1, 1].set_xlim(-3, 3)
axes[1, 1].set_ylim(-1, 3)
axes[1, 1].set_title('Arc Length: L = ∫√[(dx/dt)² + (dy/dt)²] dt', fontsize=13, fontweight='bold')
axes[1, 1].legend()

plt.tight_layout()
plt.show()

# Calculate arc length for semicircle
print("Arc Length Calculation (semicircle):")
print(f"  x = 2cos(t), y = 2sin(t), 0 ≤ t ≤ π")
print(f"  dx/dt = -2sin(t), dy/dt = 2cos(t)")
print(f"  (dx/dt)² + (dy/dt)² = 4sin²(t) + 4cos²(t) = 4")
print(f"  L = ∫₀^π √4 dt = 2∫₀^π dt = 2π")
print(f"  ✓ This matches: circumference of circle/2 = 2πr/2 = 2π (r=2)")
```

## 6. Vectors in Geometry

Vectors are **essential** for analytic geometry and calculus!

### Vector Notation
$$\vec{v} = \langle v_1, v_2 \rangle \text{ or } v_1\vec{i} + v_2\vec{j}$$

In 3D:
$$\vec{v} = \langle v_1, v_2, v_3 \rangle \text{ or } v_1\vec{i} + v_2\vec{j} + v_3\vec{k}$$

### Key Operations

**Magnitude**:
$$|\vec{v}| = \sqrt{v_1^2 + v_2^2 + v_3^2}$$

**Dot Product** (measures alignment):
$$\vec{a} \cdot \vec{b} = a_1b_1 + a_2b_2 + a_3b_3 = |\vec{a}||\vec{b}|\cos\theta$$

**Cross Product** (3D only, perpendicular vector):
$$\vec{a} \times \vec{b} = \langle a_2b_3 - a_3b_2, a_3b_1 - a_1b_3, a_1b_2 - a_2b_1 \rangle$$

```python
# Vector operations
fig = plt.figure(figsize=(18, 6))

# Plot 1: Vector addition and magnitude
ax1 = fig.add_subplot(131)
v1 = np.array([3, 2])
v2 = np.array([1, 3])
v_sum = v1 + v2

# Draw vectors
ax1.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue',
          width=0.01, label=f'v₁ = {v1}')
ax1.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='red',
          width=0.01, label=f'v₂ = {v2}')
ax1.quiver(0, 0, v_sum[0], v_sum[1], angles='xy', scale_units='xy', scale=1, color='green',
          width=0.015, label=f'v₁ + v₂ = {v_sum}')

# Parallelogram rule
ax1.quiver(v1[0], v1[1], v2[0], v2[1], angles='xy', scale_units='xy', scale=1,
          color='red', width=0.005, alpha=0.3, linestyle='dashed')
ax1.quiver(v2[0], v2[1], v1[0], v1[1], angles='xy', scale_units='xy', scale=1,
          color='blue', width=0.005, alpha=0.3, linestyle='dashed')

ax1.axhline(y=0, color='k', linewidth=0.5)
ax1.axvline(x=0, color='k', linewidth=0.5)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(-1, 6)
ax1.set_ylim(-1, 6)
ax1.set_aspect('equal')
ax1.set_title('Vector Addition\n(Parallelogram Rule)', fontsize=13, fontweight='bold')
ax1.legend()

# Plot 2: Dot product (projection)
ax2 = fig.add_subplot(132)
a_vec = np.array([4, 1])
b_vec = np.array([2, 3])

ax2.quiver(0, 0, a_vec[0], a_vec[1], angles='xy', scale_units='xy', scale=1, color='blue',
          width=0.015, label=f'a = {a_vec}')
ax2.quiver(0, 0, b_vec[0], b_vec[1], angles='xy', scale_units='xy', scale=1, color='red',
          width=0.015, label=f'b = {b_vec}')

# Projection of b onto a
dot_prod = np.dot(a_vec, b_vec)
a_mag = np.linalg.norm(a_vec)
proj_scalar = dot_prod / a_mag
proj_vec = (dot_prod / (a_mag**2)) * a_vec

ax2.quiver(0, 0, proj_vec[0], proj_vec[1], angles='xy', scale_units='xy', scale=1,
          color='green', width=0.012, label=f'proj_a(b)')
ax2.plot([b_vec[0], proj_vec[0]], [b_vec[1], proj_vec[1]], 'k--', linewidth=1, alpha=0.5)

# Angle
angle = np.arccos(dot_prod / (np.linalg.norm(a_vec) * np.linalg.norm(b_vec)))
ax2.text(1, 0.3, f'θ = {np.degrees(angle):.1f}°', fontsize=11,
        bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

ax2.axhline(y=0, color='k', linewidth=0.5)
ax2.axvline(x=0, color='k', linewidth=0.5)
ax2.grid(True, alpha=0.3)
ax2.set_xlim(-1, 5)
ax2.set_ylim(-1, 4)
ax2.set_aspect('equal')
ax2.set_title(f'Dot Product\na·b = {dot_prod} = |a||b|cosθ', fontsize=13, fontweight='bold')
ax2.legend()

# Plot 3: Cross product in 3D
ax3 = fig.add_subplot(133, projection='3d')
u_vec = np.array([2, 0, 0])
v_vec = np.array([0, 3, 0])
cross = np.cross(u_vec, v_vec)

ax3.quiver(0, 0, 0, *u_vec, color='blue', arrow_length_ratio=0.2, linewidth=3, label='u')
ax3.quiver(0, 0, 0, *v_vec, color='red', arrow_length_ratio=0.2, linewidth=3, label='v')
ax3.quiver(0, 0, 0, *cross, color='green', arrow_length_ratio=0.2, linewidth=3, 
          label=f'u × v = {cross}')

ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_zlabel('Z')
ax3.set_title('Cross Product\nPerpendicular to both!', fontsize=13, fontweight='bold')
ax3.legend()
ax3.set_xlim([0, 3])
ax3.set_ylim([0, 3])
ax3.set_zlim([0, 8])

plt.tight_layout()
plt.show()

print("Vector Operations Summary:")
print(f"\nMagnitudes:")
print(f"  |v₁| = {np.linalg.norm(v1):.3f}")
print(f"  |v₂| = {np.linalg.norm(v2):.3f}")
print(f"\nDot Product:")
print(f"  a·b = {dot_prod}")
print(f"  |a| = {np.linalg.norm(a_vec):.3f}, |b| = {np.linalg.norm(b_vec):.3f}")
print(f"  cosθ = (a·b)/(|a||b|) = {dot_prod/(np.linalg.norm(a_vec)*np.linalg.norm(b_vec)):.3f}")
print(f"  θ = {np.degrees(angle):.2f}°")
print(f"\nCross Product:")
print(f"  u × v = {cross}")
print(f"  |u × v| = {np.linalg.norm(cross):.3f} = |u||v|sinθ (area of parallelogram)")
```

## 7. Applications: Optimization with Calculus

Let's apply calculus to geometric optimization problems!

### Problem: Maximum Area Rectangle in Ellipse

Find the rectangle with maximum area that can be inscribed in an ellipse.

```python
# Optimization: Max area rectangle in ellipse
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Ellipse parameters
a_ell = 4
b_ell = 2

# Draw ellipse
theta_plot = np.linspace(0, 2*np.pi, 1000)
x_ell_plot = a_ell * np.cos(theta_plot)
y_ell_plot = b_ell * np.sin(theta_plot)

ax1.plot(x_ell_plot, y_ell_plot, 'b-', linewidth=3, label=f'Ellipse: x²/{a_ell}² + y²/{b_ell}² = 1')

# Optimal rectangle (found using calculus)
# Rectangle corners at (±x, ±y) where x = a/√2, y = b/√2
x_opt = a_ell / np.sqrt(2)
y_opt = b_ell / np.sqrt(2)
area_opt = 4 * x_opt * y_opt

# Draw optimal rectangle
rect_x = [-x_opt, x_opt, x_opt, -x_opt, -x_opt]
rect_y = [-y_opt, -y_opt, y_opt, y_opt, -y_opt]
ax1.plot(rect_x, rect_y, 'r-', linewidth=3, label=f'Max Area = {area_opt:.2f}')
ax1.fill(rect_x, rect_y, alpha=0.2, color='red')

# Mark key points
ax1.plot([x_opt, -x_opt, -x_opt, x_opt], [y_opt, y_opt, -y_opt, -y_opt], 'ro', markersize=10)

ax1.axhline(y=0, color='k', linewidth=0.5)
ax1.axvline(x=0, color='k', linewidth=0.5)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(-5, 5)
ax1.set_ylim(-3, 3)
ax1.set_aspect('equal')
ax1.set_title('Maximum Area Rectangle in Ellipse', fontsize=14, fontweight='bold')
ax1.legend()

# Plot 2: Show how area varies with x
x_range = np.linspace(0.1, a_ell*0.99, 1000)
y_range = b_ell * np.sqrt(1 - (x_range/a_ell)**2)
area_range = 4 * x_range * y_range

ax2.plot(x_range, area_range, 'b-', linewidth=3, label='Area(x) = 4xy')
ax2.plot(x_opt, area_opt, 'ro', markersize=15, label=f'Maximum at x = {x_opt:.2f}')
ax2.axvline(x=x_opt, color='r', linestyle='--', alpha=0.5)
ax2.axhline(y=area_opt, color='r', linestyle='--', alpha=0.5)

ax2.grid(True, alpha=0.3)
ax2.set_xlabel('x (half-width of rectangle)', fontsize=12)
ax2.set_ylabel('Area', fontsize=12)
ax2.set_title('Area Function\n(Maximize using dA/dx = 0)', fontsize=14, fontweight='bold')
ax2.legend()

plt.tight_layout()
plt.show()

print("Optimization Problem Solution:")
print(f"\nGiven: Ellipse x²/{a_ell}² + y²/{b_ell}² = 1")
print(f"Find: Rectangle with vertices (±x, ±y) that maximizes area")
print(f"\nSolution using Calculus:")
print(f"  1. Constraint: y = {b_ell}√(1 - x²/{a_ell}²)  [from ellipse equation]")
print(f"  2. Area: A = 4xy = 4x · {b_ell}√(1 - x²/{a_ell}²)")
print(f"  3. Take derivative: dA/dx = 0")
print(f"  4. Solve: x = {a_ell}/√2 = {x_opt:.3f}")
print(f"  5. Then: y = {b_ell}/√2 = {y_opt:.3f}")
print(f"\nMaximum Area = {area_opt:.3f}")
print(f"\n(Compare to ellipse area πab = π·{a_ell}·{b_ell} = {np.pi * a_ell * b_ell:.3f})")
```

### Problem 1: Convert Coordinates

Convert the point $(x, y) = (-3, 4)$ to polar coordinates.

```python
print("Problem 1: Convert (-3, 4) to polar coordinates")
print("\nSolution:")
x_prob, y_prob = -3, 4
r_prob = np.sqrt(x_prob**2 + y_prob**2)
theta_prob = np.arctan2(y_prob, x_prob)  # arctan2 handles quadrants correctly

print(f"  r = √(x² + y²) = √((-3)² + 4²) = √(9 + 16) = √25 = {r_prob}")
print(f"  θ = arctan(y/x) = arctan(4/(-3))")
print(f"  Since x < 0 and y > 0, we're in Quadrant II")
print(f"  θ = {theta_prob:.4f} rad = {np.degrees(theta_prob):.2f}°")
print(f"\nAnswer: (r, θ) = ({r_prob}, {np.degrees(theta_prob):.2f}°) or ({r_prob}, {theta_prob:.4f} rad)")

# Verify
x_verify = r_prob * np.cos(theta_prob)
y_verify = r_prob * np.sin(theta_prob)
print(f"\nVerification:")
print(f"  x = r·cos(θ) = {r_prob}·cos({theta_prob:.4f}) = {x_verify:.3f}")
print(f"  y = r·sin(θ) = {r_prob}·sin({theta_prob:.4f}) = {y_verify:.3f}")
print(f"  ✓ Matches original point!")
```

### Problem 2: Find Distance Between Points in 3D

Find the distance between $P_1 = (1, 2, 3)$ and $P_2 = (4, 0, -1)$.

```python
print("Problem 2: Distance between (1, 2, 3) and (4, 0, -1)")
print("\nSolution:")
P1 = np.array([1, 2, 3])
P2 = np.array([4, 0, -1])

print(f"  d = √[(x₂-x₁)² + (y₂-y₁)² + (z₂-z₁)²]")
print(f"    = √[(4-1)² + (0-2)² + (-1-3)²]")
print(f"    = √[3² + (-2)² + (-4)²]")
print(f"    = √[9 + 4 + 16]")
print(f"    = √29")

distance = np.linalg.norm(P2 - P1)
print(f"\nAnswer: d = {distance:.4f}")
```

### Problem 3: Find Equation of Circle

Find the equation of the circle with center $(2, -3)$ passing through the point $(5, 1)$.

```python
print("Problem 3: Circle with center (2, -3) through point (5, 1)")
print("\nSolution:")
h_center, k_center = 2, -3
x_point, y_point = 5, 1

print(f"  Step 1: Find radius (distance from center to point)")
r_circle = np.sqrt((x_point - h_center)**2 + (y_point - k_center)**2)
print(f"          r = √[(5-2)² + (1-(-3))²] = √[9 + 16] = √25 = {r_circle}")

print(f"\n  Step 2: Write equation in standard form")
print(f"          (x - h)² + (y - k)² = r²")
print(f"          (x - {h_center})² + (y - ({k_center}))² = {r_circle}²")
print(f"\nAnswer: (x - 2)² + (y + 3)² = 25")

# Verify point is on circle
check = (x_point - h_center)**2 + (y_point - k_center)**2
print(f"\nVerification: ({x_point}-2)² + ({y_point}+3)² = {check} = 25 ✓")
```

## 9. Key Takeaways

### Coordinate Systems
- **Cartesian**: $(x, y)$ or $(x, y, z)$ - standard rectangular coordinates
- **Polar**: $(r, \theta)$ - useful for circles, spirals, rotation
- **Conversion**: $x = r\cos\theta$, $y = r\sin\theta$

### Conic Sections (MEMORIZE!)
- **Circle**: $(x-h)^2 + (y-k)^2 = r^2$
- **Ellipse**: $\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1$ (sum of distances to foci = constant)
- **Parabola**: $x^2 = 4py$ (distance to focus = distance to directrix)
- **Hyperbola**: $\frac{x^2}{a^2} - \frac{y^2}{b^2} = 1$ (difference of distances to foci = constant)

### Parametric Equations
- Express curves as $x = f(t)$, $y = g(t)$
- **Arc length**: $L = \int \sqrt{(\frac{dx}{dt})^2 + (\frac{dy}{dt})^2} dt$
- Natural for motion and curves that fail vertical line test

### Vectors
- **Dot product**: $\vec{a} \cdot \vec{b} = |\vec{a}||\vec{b}|\cos\theta$ (alignment)
- **Cross product**: $\vec{a} \times \vec{b}$ gives perpendicular vector (3D only)
- Essential for multivariable calculus!

### Calculus Connections
- **Tangent lines**: Use implicit differentiation
- **Arc length**: Integration of parametric derivatives
- **Area**: Integration over regions
- **Optimization**: Find max/min using derivatives
