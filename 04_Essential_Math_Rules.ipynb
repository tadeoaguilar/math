# Essential Math Rules for Calculus

## Why These Rules Matter

Calculus builds on algebra and trigonometry. You'll constantly need to:
- **Simplify expressions** before differentiating or integrating
- **Factor polynomials** to find critical points
- **Use trig identities** to simplify derivatives and integrals
- **Complete the square** for integration techniques
- **Manipulate fractions** for partial fraction decomposition

**Goal**: Have these rules at your fingertips so they don't slow down your calculus work!

## 1. Algebraic Fundamentals

### Factoring Patterns

**Difference of squares**:
$$a^2 - b^2 = (a-b)(a+b)$$

**Perfect square trinomials**:
$$a^2 + 2ab + b^2 = (a+b)^2$$
$$a^2 - 2ab + b^2 = (a-b)^2$$

**Difference/Sum of cubes**:
$$a^3 - b^3 = (a-b)(a^2 + ab + b^2)$$
$$a^3 + b^3 = (a+b)(a^2 - ab + b^2)$$

**General quadratic**:
$$ax^2 + bx + c = a(x - r_1)(x - r_2)$$
where $r_1, r_2 = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$

```python
print("=" * 70)
print("FACTORING EXAMPLES")
print("=" * 70)

print("\n1. DIFFERENCE OF SQUARES: a² - b² = (a-b)(a+b)")
examples = [
    ("x² - 9", "(x-3)(x+3)"),
    ("4x² - 25", "(2x-5)(2x+5)"),
    ("x² - 1", "(x-1)(x+1)"),
]
for expr, factored in examples:
    print(f"   {expr:15} = {factored}")

print("\n2. PERFECT SQUARE TRINOMIALS: a² ± 2ab + b² = (a ± b)²")
examples = [
    ("x² + 6x + 9", "(x+3)²"),
    ("x² - 10x + 25", "(x-5)²"),
    ("4x² + 12x + 9", "(2x+3)²"),
]
for expr, factored in examples:
    print(f"   {expr:20} = {factored}")

print("\n3. SUM/DIFFERENCE OF CUBES")
print("   a³ - b³ = (a-b)(a² + ab + b²)")
print("   a³ + b³ = (a+b)(a² - ab + b²)")
examples = [
    ("x³ - 8", "(x-2)(x² + 2x + 4)"),
    ("x³ + 27", "(x+3)(x² - 3x + 9)"),
]
for expr, factored in examples:
    print(f"   {expr:15} = {factored}")

print("\n4. GENERAL QUADRATICS")
examples = [
    ("x² + 5x + 6", "(x+2)(x+3)"),
    ("x² - 7x + 12", "(x-3)(x-4)"),
    ("2x² + 7x + 3", "(2x+1)(x+3)"),
    ("x² - 4x - 5", "(x-5)(x+1)"),
]
for expr, factored in examples:
    print(f"   {expr:20} = {factored}")

print("\n" + "=" * 70)
```

### Completing the Square

Transform $ax^2 + bx + c$ into vertex form $a(x-h)^2 + k$:

**Steps**:
1. Factor out $a$ from $x^2$ and $x$ terms (if $a \neq 1$)
2. Take half of the $x$ coefficient, square it
3. Add and subtract this value inside the parentheses
4. Simplify

**Example**: $2x^2 + 12x + 10$
1. $2(x^2 + 6x) + 10$
2. Half of 6 is 3, squared is 9
3. $2(x^2 + 6x + 9 - 9) + 10$
4. $2(x + 3)^2 - 18 + 10 = 2(x+3)^2 - 8$

```python
# Visualize completing the square
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Example: x² + 6x + 5
x = np.linspace(-8, 2, 1000)
y_standard = x**2 + 6*x + 5
y_vertex = (x + 3)**2 - 4  # Completed square form

axes[0].plot(x, y_standard, 'b-', linewidth=3, label='y = x² + 6x + 5')
axes[0].plot(-3, -4, 'ro', markersize=15, label='Vertex: (-3, -4)')
axes[0].axhline(y=0, color='gray', linewidth=1, alpha=0.5)
axes[0].axvline(x=-3, color='r', linestyle='--', linewidth=2, alpha=0.5, label='Axis of symmetry')
axes[0].set_xlabel('x', fontsize=13)
axes[0].set_ylabel('y', fontsize=13)
axes[0].set_title('Standard Form: y = x² + 6x + 5\nVertex Form: y = (x+3)² - 4', 
                   fontsize=14, fontweight='bold')
axes[0].legend(fontsize=12)
axes[0].grid(True, alpha=0.3)
axes[0].set_ylim(-6, 10)

# Show the completing the square process
axes[1].text(0.5, 0.95, 'Completing the Square', transform=axes[1].transAxes,
             fontsize=16, fontweight='bold', ha='center', va='top')
axes[1].text(0.5, 0.85, 'Convert: x² + 6x + 5 to vertex form', transform=axes[1].transAxes,
             fontsize=13, ha='center', va='top')

steps = [
    "Step 1: Identify the x² and x terms",
    "        x² + 6x + 5",
    "",
    "Step 2: Take half of 6, square it: (6/2)² = 9",
    "",
    "Step 3: Add and subtract 9",
    "        x² + 6x + 9 - 9 + 5",
    "",
    "Step 4: Factor the perfect square",
    "        (x + 3)² - 9 + 5",
    "",
    "Step 5: Simplify",
    "        (x + 3)² - 4",
    "",
    "Vertex: h = -3, k = -4",
    "Vertex form: y = (x + 3)² - 4"
]

y_pos = 0.75
for step in steps:
    if step.startswith('Step'):
        axes[1].text(0.1, y_pos, step, transform=axes[1].transAxes,
                     fontsize=11, fontweight='bold', va='top', family='monospace')
    else:
        axes[1].text(0.1, y_pos, step, transform=axes[1].transAxes,
                     fontsize=11, va='top', family='monospace')
    y_pos -= 0.045

axes[1].axis('off')

plt.tight_layout()
plt.show()

print("Why Completing the Square Matters in Calculus:")
print("  • Finding vertex of parabolas (optimization problems)")
print("  • Integration techniques (especially trigonometric substitution)")
print("  • Analyzing conic sections")
print("  • Solving differential equations")
```

### Essential Trigonometric Identities

#### Pythagorean Identities
$$\sin^2\theta + \cos^2\theta = 1$$
$$1 + \tan^2\theta = \sec^2\theta$$
$$1 + \cot^2\theta = \csc^2\theta$$

#### Angle Sum/Difference
$$\sin(A \pm B) = \sin A \cos B \pm \cos A \sin B$$
$$\cos(A \pm B) = \cos A \cos B \mp \sin A \sin B$$
$$\tan(A \pm B) = \frac{\tan A \pm \tan B}{1 \mp \tan A \tan B}$$

#### Double Angle
$$\sin(2\theta) = 2\sin\theta\cos\theta$$
$$\cos(2\theta) = \cos^2\theta - \sin^2\theta = 2\cos^2\theta - 1 = 1 - 2\sin^2\theta$$
$$\tan(2\theta) = \frac{2\tan\theta}{1 - \tan^2\theta}$$

#### Half Angle (for integration!)
$$\sin^2\theta = \frac{1 - \cos(2\theta)}{2}$$
$$\cos^2\theta = \frac{1 + \cos(2\theta)}{2}$$

#### Product-to-Sum (for integration!)
$$\sin A \cos B = \frac{1}{2}[\sin(A+B) + \sin(A-B)]$$
$$\cos A \cos B = \frac{1}{2}[\cos(A+B) + \cos(A-B)]$$
$$\sin A \sin B = \frac{1}{2}[\cos(A-B) - \cos(A+B)]$$

```python
# Verify trig identities numerically
print("=" * 70)
print("VERIFYING TRIGONOMETRIC IDENTITIES")
print("=" * 70)

theta = np.pi / 6  # 30 degrees

print(f"\nUsing θ = π/6 (30°)")
print("-" * 70)

# Pythagorean identity
print("\n1. PYTHAGOREAN IDENTITY: sin²θ + cos²θ = 1")
lhs = np.sin(theta)**2 + np.cos(theta)**2
print(f"   sin²(π/6) + cos²(π/6) = {np.sin(theta)**2:.6f} + {np.cos(theta)**2:.6f} = {lhs:.6f}")
print(f"   ✓ Equals 1.000000")

# Double angle
print("\n2. DOUBLE ANGLE: sin(2θ) = 2sin(θ)cos(θ)")
lhs = np.sin(2*theta)
rhs = 2 * np.sin(theta) * np.cos(theta)
print(f"   sin(2·π/6) = sin(π/3) = {lhs:.6f}")
print(f"   2sin(π/6)cos(π/6) = 2·{np.sin(theta):.6f}·{np.cos(theta):.6f} = {rhs:.6f}")
print(f"   ✓ Both equal {lhs:.6f}")

# cos double angle
print("\n3. DOUBLE ANGLE: cos(2θ) = cos²θ - sin²θ")
lhs = np.cos(2*theta)
rhs1 = np.cos(theta)**2 - np.sin(theta)**2
rhs2 = 2*np.cos(theta)**2 - 1
rhs3 = 1 - 2*np.sin(theta)**2
print(f"   cos(π/3) = {lhs:.6f}")
print(f"   cos²(π/6) - sin²(π/6) = {rhs1:.6f}")
print(f"   2cos²(π/6) - 1 = {rhs2:.6f}")
print(f"   1 - 2sin²(π/6) = {rhs3:.6f}")
print(f"   ✓ All forms equal {lhs:.6f}")

# Angle sum
A, B = np.pi/6, np.pi/4
print("\n4. ANGLE SUM: sin(A+B) = sinA·cosB + cosA·sinB")
print(f"   Using A = π/6, B = π/4")
lhs = np.sin(A + B)
rhs = np.sin(A)*np.cos(B) + np.cos(A)*np.sin(B)
print(f"   sin(π/6 + π/4) = sin(5π/12) = {lhs:.6f}")
print(f"   sin(π/6)cos(π/4) + cos(π/6)sin(π/4) = {rhs:.6f}")
print(f"   ✓ Both equal {lhs:.6f}")

# Power reduction
print("\n5. POWER REDUCTION: sin²θ = (1 - cos(2θ))/2")
lhs = np.sin(theta)**2
rhs = (1 - np.cos(2*theta)) / 2
print(f"   sin²(π/6) = {lhs:.6f}")
print(f"   (1 - cos(π/3))/2 = {rhs:.6f}")
print(f"   ✓ Both equal {lhs:.6f}")

print("\n" + "=" * 70)
```

## 3. Rational Expressions and Partial Fractions

### Simplifying Rational Expressions

**Factor and cancel**:
$$\frac{x^2 - 4}{x^2 - 5x + 6} = \frac{(x-2)(x+2)}{(x-2)(x-3)} = \frac{x+2}{x-3} \quad (x \neq 2)$$

### Adding/Subtracting Rational Expressions

**Common denominator**:
$$\frac{2}{x+1} + \frac{3}{x-1} = \frac{2(x-1) + 3(x+1)}{(x+1)(x-1)} = \frac{5x+1}{x^2-1}$$

### Partial Fraction Decomposition (Crucial for Integration!)

Break apart a complex fraction into simpler pieces:

$$\frac{5x+1}{(x+1)(x-1)} = \frac{A}{x+1} + \frac{B}{x-1}$$

**Types**:
1. **Distinct linear factors**: $\frac{P(x)}{(x-a)(x-b)} = \frac{A}{x-a} + \frac{B}{x-b}$

2. **Repeated linear factors**: $\frac{P(x)}{(x-a)^2} = \frac{A}{x-a} + \frac{B}{(x-a)^2}$

3. **Irreducible quadratic**: $\frac{P(x)}{(x^2+bx+c)} = \frac{Ax+B}{x^2+bx+c}$

```python
print("=" * 70)
print("PARTIAL FRACTION DECOMPOSITION")
print("=" * 70)

print("\nEXAMPLE 1: Distinct Linear Factors")
print("-" * 70)
print("Decompose: (5x + 1)/((x+1)(x-1))")
print("")
print("Step 1: Set up the form")
print("        (5x + 1)/((x+1)(x-1)) = A/(x+1) + B/(x-1)")
print("")
print("Step 2: Clear denominators")
print("        5x + 1 = A(x-1) + B(x+1)")
print("")
print("Step 3: Solve for A and B using strategic x values")
print("        Let x = 1:  5(1) + 1 = A(0) + B(2)")
print("                    6 = 2B  →  B = 3")
print("        Let x = -1: 5(-1) + 1 = A(-2) + B(0)")
print("                    -4 = -2A  →  A = 2")
print("")
print("Step 4: Final decomposition")
print("        (5x + 1)/((x+1)(x-1)) = 2/(x+1) + 3/(x-1)")
print("")
print("Verification:")
# Verify numerically
x_test = 2
original = (5*x_test + 1) / ((x_test + 1) * (x_test - 1))
decomposed = 2/(x_test + 1) + 3/(x_test - 1)
print(f"  At x = {x_test}:")
print(f"    Original: {original:.6f}")
print(f"    Decomposed: {decomposed:.6f}")
print(f"    ✓ Match!")

print("\n" + "=" * 70)
print("\nEXAMPLE 2: Repeated Linear Factor")
print("-" * 70)
print("Decompose: (3x + 5)/(x(x-1)²)")
print("")
print("Step 1: Set up the form (note the repeated factor!)")
print("        (3x + 5)/(x(x-1)²) = A/x + B/(x-1) + C/(x-1)²")
print("")
print("Step 2: Clear denominators")
print("        3x + 5 = A(x-1)² + Bx(x-1) + Cx")
print("")
print("Step 3: Solve using strategic x values and expansion")
print("        Let x = 0:  5 = A(1)  →  A = 5")
print("        Let x = 1:  8 = C(1)  →  C = 8")
print("        Expand and compare coefficients for B = -5")
print("")
print("Step 4: Final decomposition")
print("        (3x + 5)/(x(x-1)²) = 5/x - 5/(x-1) + 8/(x-1)²")

print("\n" + "=" * 70)
print("\nWhy Partial Fractions Matter for Calculus:")
print("-" * 70)
print("  • Essential for integrating rational functions")
print("  • ∫ 1/(x-a) dx = ln|x-a| + C  (easy to integrate!)")
print("  • Complex fractions become sums of simple logarithms")
print("  • One of the most important integration techniques")
print("=" * 70)
```

| One-to-one | Each output from one input | $e^x, x^3$ |
| Increasing | $x_1 < x_2 \Rightarrow f(x_1) < f(x_2)$ | $e^x, \ln x$ |
| Decreasing | $x_1 < x_2 \Rightarrow f(x_1) > f(x_2)$ | $-x, e^{-x}$ |

```python
# Create a comprehensive reference visualization
fig = plt.figure(figsize=(16, 10))
gs = fig.add_gridspec(3, 3, hspace=0.4, wspace=0.3)

# Common functions
x = np.linspace(-3, 3, 1000)

# Polynomial
ax1 = fig.add_subplot(gs[0, 0])
ax1.plot(x, x**2, 'b-', linewidth=2)
ax1.set_title('Quadratic: x²', fontsize=11, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.set_ylim(-1, 9)

# Cubic
ax2 = fig.add_subplot(gs[0, 1])
ax2.plot(x, x**3, 'r-', linewidth=2)
ax2.set_title('Cubic: x³', fontsize=11, fontweight='bold')
ax2.grid(True, alpha=0.3)

# Square root
ax3 = fig.add_subplot(gs[0, 2])
x_pos = np.linspace(0, 9, 1000)
ax3.plot(x_pos, np.sqrt(x_pos), 'g-', linewidth=2)
ax3.set_title('Square Root: √x', fontsize=11, fontweight='bold')
ax3.grid(True, alpha=0.3)

# Exponential
ax4 = fig.add_subplot(gs[1, 0])
ax4.plot(x, np.exp(x), 'b-', linewidth=2)
ax4.set_title('Exponential: eˣ', fontsize=11, fontweight='bold')
ax4.grid(True, alpha=0.3)
ax4.set_ylim(-1, 10)

# Natural log
ax5 = fig.add_subplot(gs[1, 1])
ax5.plot(x_pos, np.log(x_pos), 'r-', linewidth=2)
ax5.axvline(x=0, color='gray', linestyle='--', alpha=0.5)
ax5.set_title('Natural Log: ln(x)', fontsize=11, fontweight='bold')
ax5.grid(True, alpha=0.3)

# Sine
ax6 = fig.add_subplot(gs[1, 2])
x_trig = np.linspace(-2*np.pi, 2*np.pi, 1000)
ax6.plot(x_trig, np.sin(x_trig), 'g-', linewidth=2)
ax6.set_title('Sine: sin(x)', fontsize=11, fontweight='bold')
ax6.grid(True, alpha=0.3)

# Cosine
ax7 = fig.add_subplot(gs[2, 0])
ax7.plot(x_trig, np.cos(x_trig), 'b-', linewidth=2)
ax7.set_title('Cosine: cos(x)', fontsize=11, fontweight='bold')
ax7.grid(True, alpha=0.3)

# Tangent
ax8 = fig.add_subplot(gs[2, 1])
x_tan = np.linspace(-1.4, 1.4, 1000)
ax8.plot(x_tan, np.tan(x_tan), 'r-', linewidth=2)
ax8.axvline(x=np.pi/2, color='gray', linestyle='--', alpha=0.5)
ax8.axvline(x=-np.pi/2, color='gray', linestyle='--', alpha=0.5)
ax8.set_title('Tangent: tan(x)', fontsize=11, fontweight='bold')
ax8.grid(True, alpha=0.3)
ax8.set_ylim(-5, 5)

# Absolute value
ax9 = fig.add_subplot(gs[2, 2])
ax9.plot(x, np.abs(x), 'g-', linewidth=2)
ax9.set_title('Absolute Value: |x|', fontsize=11, fontweight='bold')
ax9.grid(True, alpha=0.3)

for ax in [ax1, ax2, ax3, ax4, ax5, ax6, ax7, ax8, ax9]:
    ax.axhline(y=0, color='black', linewidth=0.5)
    ax.axvline(x=0, color='black', linewidth=0.5)

plt.suptitle('Essential Functions Quick Reference', fontsize=16, fontweight='bold', y=0.995)
plt.show()
```
