# Trigonometry: Essential Foundation for Calculus

## Why Trigonometry Matters for Calculus

Trigonometry is **crucial** for calculus! Here's why:

- **Derivatives**: You'll need to know derivatives of sin, cos, tan, and their inverses
- **Integration**: Many integrals require trig substitution techniques
- **Trig Identities**: Essential for simplifying expressions before differentiating/integrating
- **Applications**: Physics (waves, oscillations), engineering (signals), geometry
- **Special Limits**: The limit $\lim_{x \to 0} \frac{\sin x}{x} = 1$ is fundamental

**The Big Idea**: Trigonometry describes relationships in triangles and circular motion - it's the mathematics of angles and periodic phenomena.

### Why the Circle? The Power of Generalization

**The Problem with Triangles**: Right triangles only work for angles between 0° and 90°. But what about 120°? Or 270°? Or even -30°?

**The Solution - The Circle**: By placing our triangle inside a circle with radius = 1, we can define sin and cos for **ANY angle**:

1. **Start at the positive x-axis** (angle = 0°)
2. **Rotate counter-clockwise** by angle θ
3. **Drop a perpendicular** to the x-axis
4. **Read the coordinates**: The point where you end up is (cos θ, sin θ)

This works for:
- **Angles > 90°**: Keep rotating around the circle
- **Angles > 360°**: Just keep going around - you'll wrap back
- **Negative angles**: Rotate clockwise instead

### The Unit Circle - Why Radius = 1?

When the radius = 1:
- The hypotenuse is always 1
- sin(θ) = opposite/1 = opposite = **y-coordinate**
- cos(θ) = adjacent/1 = adjacent = **x-coordinate**

This makes everything cleaner and is why we call it the "**unit**" circle (unit = 1).

Now let's see this in action with different angles!

## 0. What ARE Sine, Cosine, and Tangent? (Building from First Principles)

Before diving into formulas, let's understand **what these functions actually mean**. We'll start with the most basic building block: a right triangle.

### The Right Triangle - Where It All Begins

A **right triangle** has:
- One 90° angle (the right angle)
- Two other angles that add up to 90°
- Three sides

**Key Insight**: For any specific angle θ in a right triangle, the **ratios** between the sides are always the same, no matter how big or small the triangle is!

This is the foundation of all trigonometry.

### The Three Basic Ratios (SOH-CAH-TOA)

For any angle θ in a right triangle:

$$\sin(\theta) = \frac{\text{opposite side}}{\text{hypotenuse}} \quad \text{(SOH)}$$

$$\cos(\theta) = \frac{\text{adjacent side}}{\text{hypotenuse}} \quad \text{(CAH)}$$

$$\tan(\theta) = \frac{\text{opposite side}}{\text{adjacent side}} \quad \text{(TOA)}$$

**Mnemonic**: **SOH-CAH-TOA** (Some Old Hippie Caught Another Hippie Trippin' On Acid)

Let's visualize this!

## 2. The Unit Circle

The **unit circle** is a circle with radius 1 centered at the origin.

### Equation
$$x^2 + y^2 = 1$$

### Connection to Trigonometry

For any angle $\theta$, the point on the unit circle is:
$$P(\theta) = (\cos \theta, \sin \theta)$$

This is the **fundamental definition** of sine and cosine!

### Key Properties
- $-1 \leq \cos \theta \leq 1$
- $-1 \leq \sin \theta \leq 1$
- $\cos^2 \theta + \sin^2 \theta = 1$ (Pythagorean identity!)

```python
# The Unit Circle with all special angles
fig, ax = plt.subplots(figsize=(14, 14))

# Draw unit circle
circle = Circle((0, 0), 1, fill=False, linewidth=3, edgecolor='black')
ax.add_patch(circle)

# Key angles with exact values
special_angles = [
    (0, '0', (1, 0)),
    (np.pi/6, 'π/6\n30°', (np.sqrt(3)/2, 1/2)),
    (np.pi/4, 'π/4\n45°', (np.sqrt(2)/2, np.sqrt(2)/2)),
    (np.pi/3, 'π/3\n60°', (1/2, np.sqrt(3)/2)),
    (np.pi/2, 'π/2\n90°', (0, 1)),
    (2*np.pi/3, '2π/3\n120°', (-1/2, np.sqrt(3)/2)),
    (3*np.pi/4, '3π/4\n135°', (-np.sqrt(2)/2, np.sqrt(2)/2)),
    (5*np.pi/6, '5π/6\n150°', (-np.sqrt(3)/2, 1/2)),
    (np.pi, 'π\n180°', (-1, 0)),
    (7*np.pi/6, '7π/6\n210°', (-np.sqrt(3)/2, -1/2)),
    (5*np.pi/4, '5π/4\n225°', (-np.sqrt(2)/2, -np.sqrt(2)/2)),
    (4*np.pi/3, '4π/3\n240°', (-1/2, -np.sqrt(3)/2)),
    (3*np.pi/2, '3π/2\n270°', (0, -1)),
    (5*np.pi/3, '5π/3\n300°', (1/2, -np.sqrt(3)/2)),
    (7*np.pi/4, '7π/4\n315°', (np.sqrt(2)/2, -np.sqrt(2)/2)),
    (11*np.pi/6, '11π/6\n330°', (np.sqrt(3)/2, -1/2)),
]

for angle, label, (cos_val, sin_val) in special_angles:
    x = np.cos(angle)
    y = np.sin(angle)
    
    # Draw radius line
    ax.plot([0, x], [0, y], 'b-', linewidth=1.5, alpha=0.6)
    
    # Mark point on circle
    ax.plot(x, y, 'ro', markersize=8)
    
    # Label with angle
    label_x = x * 1.35
    label_y = y * 1.35
    ax.text(label_x, label_y, label, fontsize=9, ha='center', va='center',
           bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.6))
    
    # Add coordinates
    coord_text = f'({cos_val:.3f}, {sin_val:.3f})'
    coord_x = x * 0.65
    coord_y = y * 0.65
    ax.text(coord_x, coord_y, coord_text, fontsize=7, ha='center', va='center',
           bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))

# Add axes
ax.axhline(y=0, color='k', linewidth=2)
ax.axvline(x=0, color='k', linewidth=2)

# Add labels for quadrants
ax.text(0.7, 0.7, 'Quadrant I\nAll positive', fontsize=11, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.3))
ax.text(-0.7, 0.7, 'Quadrant II\nsin > 0, cos < 0', fontsize=11, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.3))
ax.text(-0.7, -0.7, 'Quadrant III\nAll negative', fontsize=11, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.3))
ax.text(0.7, -0.7, 'Quadrant IV\ncos > 0, sin < 0', fontsize=11, ha='center',
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.3))

ax.set_xlim(-1.7, 1.7)
ax.set_ylim(-1.7, 1.7)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_xlabel('x = cos(θ)', fontsize=13, fontweight='bold')
ax.set_ylabel('y = sin(θ)', fontsize=13, fontweight='bold')
ax.set_title('The Unit Circle: Your Best Friend in Trigonometry!', fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()

print("MEMORIZE THIS CIRCLE! It contains everything you need for basic trigonometry.")
```

## 3. The Six Trigonometric Functions

### Primary Functions (from unit circle)

$$\sin \theta = y \quad \text{(opposite/hypotenuse)}$$
$$\cos \theta = x \quad \text{(adjacent/hypotenuse)}$$
$$\tan \theta = \frac{y}{x} = \frac{\sin \theta}{\cos \theta} \quad \text{(opposite/adjacent)}$$

### Reciprocal Functions

$$\csc \theta = \frac{1}{\sin \theta}$$
$$\sec \theta = \frac{1}{\cos \theta}$$
$$\cot \theta = \frac{1}{\tan \theta} = \frac{\cos \theta}{\sin \theta}$$

### Special Values to Memorize

| θ | sin θ | cos θ | tan θ |
|---|-------|-------|-------|
| 0 | 0 | 1 | 0 |
| π/6 (30°) | 1/2 | √3/2 | √3/3 |
| π/4 (45°) | √2/2 | √2/2 | 1 |
| π/3 (60°) | √3/2 | 1/2 | √3 |
| π/2 (90°) | 1 | 0 | undefined |

```python
# Visualize how trig functions relate to the unit circle
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# Choose an angle to demonstrate
theta = np.pi/3  # 60 degrees
x_coord = np.cos(theta)
y_coord = np.sin(theta)

# Function to draw unit circle with labels
def draw_base_circle(ax, theta):
    circle = Circle((0, 0), 1, fill=False, linewidth=2, edgecolor='black')
    ax.add_patch(circle)
    ax.axhline(y=0, color='k', linewidth=1)
    ax.axvline(x=0, color='k', linewidth=1)
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    
    # Draw angle arc
    arc_theta = np.linspace(0, theta, 50)
    arc_r = 0.3
    ax.plot(arc_r * np.cos(arc_theta), arc_r * np.sin(arc_theta), 'g-', linewidth=2)
    ax.text(0.4, 0.15, 'θ', fontsize=14, color='green', fontweight='bold')

# 1. sin(θ) = y-coordinate
draw_base_circle(axes[0, 0], theta)
axes[0, 0].plot([0, x_coord], [0, y_coord], 'b-', linewidth=3, label='radius')
axes[0, 0].plot(x_coord, y_coord, 'ro', markersize=12)
axes[0, 0].plot([x_coord, x_coord], [0, y_coord], 'r-', linewidth=4, label=f'sin(θ) = {y_coord:.3f}')
axes[0, 0].text(x_coord + 0.15, y_coord/2, f'sin(θ)\n= {y_coord:.3f}', fontsize=12, 
               bbox=dict(boxstyle='round', facecolor='red', alpha=0.3))
axes[0, 0].set_title('sin(θ) = y-coordinate', fontsize=13, fontweight='bold')
axes[0, 0].legend()

# 2. cos(θ) = x-coordinate
draw_base_circle(axes[0, 1], theta)
axes[0, 1].plot([0, x_coord], [0, y_coord], 'b-', linewidth=3, label='radius')
axes[0, 1].plot(x_coord, y_coord, 'ro', markersize=12)
axes[0, 1].plot([0, x_coord], [0, 0], 'b-', linewidth=4, label=f'cos(θ) = {x_coord:.3f}')
axes[0, 1].text(x_coord/2, -0.15, f'cos(θ) = {x_coord:.3f}', fontsize=12,
               bbox=dict(boxstyle='round', facecolor='blue', alpha=0.3))
axes[0, 1].set_title('cos(θ) = x-coordinate', fontsize=13, fontweight='bold')
axes[0, 1].legend()

# 3. tan(θ) = sin/cos = slope
draw_base_circle(axes[0, 2], theta)
axes[0, 2].plot([0, x_coord], [0, y_coord], 'purple', linewidth=3, label='radius')
axes[0, 2].plot(x_coord, y_coord, 'ro', markersize=12)
# Extend line to x=1 to show tangent
tan_val = y_coord / x_coord
axes[0, 2].plot([0, 1], [0, tan_val], 'g--', linewidth=2, alpha=0.7)
axes[0, 2].plot([1, 1], [0, tan_val], 'orange', linewidth=4, label=f'tan(θ) = {tan_val:.3f}')
axes[0, 2].text(1.1, tan_val/2, f'tan(θ)\n= {tan_val:.3f}', fontsize=12,
               bbox=dict(boxstyle='round', facecolor='orange', alpha=0.3))
axes[0, 2].set_title('tan(θ) = sin/cos (tangent line)', fontsize=13, fontweight='bold')
axes[0, 2].legend()

# 4. csc(θ) = 1/sin
draw_base_circle(axes[1, 0], theta)
axes[1, 0].plot([0, x_coord], [0, y_coord], 'b-', linewidth=3)
axes[1, 0].plot(x_coord, y_coord, 'ro', markersize=12)
csc_val = 1/y_coord
axes[1, 0].plot([0, x_coord], [0, csc_val], 'purple', linewidth=2, linestyle='--', alpha=0.7)
axes[1, 0].text(0.3, csc_val + 0.1, f'csc(θ) = 1/sin(θ) = {csc_val:.3f}', fontsize=11,
               bbox=dict(boxstyle='round', facecolor='purple', alpha=0.3))
axes[1, 0].set_title('csc(θ) = 1/sin(θ)', fontsize=13, fontweight='bold')

# 5. sec(θ) = 1/cos
draw_base_circle(axes[1, 1], theta)
axes[1, 1].plot([0, x_coord], [0, y_coord], 'b-', linewidth=3)
axes[1, 1].plot(x_coord, y_coord, 'ro', markersize=12)
sec_val = 1/x_coord
axes[1, 1].plot([0, sec_val], [0, 0], 'green', linewidth=4, alpha=0.7)
axes[1, 1].text(sec_val/2, -0.2, f'sec(θ) = 1/cos(θ) = {sec_val:.3f}', fontsize=11,
               bbox=dict(boxstyle='round', facecolor='green', alpha=0.3))
axes[1, 1].set_title('sec(θ) = 1/cos(θ)', fontsize=13, fontweight='bold')

# 6. cot(θ) = 1/tan = cos/sin
draw_base_circle(axes[1, 2], theta)
axes[1, 2].plot([0, x_coord], [0, y_coord], 'b-', linewidth=3)
axes[1, 2].plot(x_coord, y_coord, 'ro', markersize=12)
cot_val = x_coord / y_coord
axes[1, 2].plot([0, cot_val], [0, 1], 'brown', linewidth=2, linestyle='--', alpha=0.7)
axes[1, 2].plot([0, cot_val], [1, 1], 'brown', linewidth=4, alpha=0.7)
axes[1, 2].text(cot_val/2, 1.15, f'cot(θ) = cos/sin = {cot_val:.3f}', fontsize=11,
               bbox=dict(boxstyle='round', facecolor='brown', alpha=0.3))
axes[1, 2].set_title('cot(θ) = cos(θ)/sin(θ)', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()

print(f"For θ = π/3 (60°):")
print(f"  sin(θ) = {np.sin(theta):.4f}")
print(f"  cos(θ) = {np.cos(theta):.4f}")
print(f"  tan(θ) = {np.tan(theta):.4f}")
print(f"  csc(θ) = {1/np.sin(theta):.4f}")
print(f"  sec(θ) = {1/np.cos(theta):.4f}")
print(f"  cot(θ) = {1/np.tan(theta):.4f}")
```

## 4. Essential Trigonometric Identities

These identities are **crucial** for calculus! You'll use them constantly.

### Pythagorean Identities
$$\sin^2 \theta + \cos^2 \theta = 1$$
$$1 + \tan^2 \theta = \sec^2 \theta$$
$$1 + \cot^2 \theta = \csc^2 \theta$$

### Even/Odd Identities
$$\sin(-\theta) = -\sin \theta \quad \text{(odd)}$$
$$\cos(-\theta) = \cos \theta \quad \text{(even)}$$
$$\tan(-\theta) = -\tan \theta \quad \text{(odd)}$$

### Sum and Difference Formulas
$$\sin(A \pm B) = \sin A \cos B \pm \cos A \sin B$$
$$\cos(A \pm B) = \cos A \cos B \mp \sin A \sin B$$
$$\tan(A \pm B) = \frac{\tan A \pm \tan B}{1 \mp \tan A \tan B}$$

### Double Angle Formulas
$$\sin(2\theta) = 2\sin \theta \cos \theta$$
$$\cos(2\theta) = \cos^2 \theta - \sin^2 \theta = 2\cos^2 \theta - 1 = 1 - 2\sin^2 \theta$$
$$\tan(2\theta) = \frac{2\tan \theta}{1 - \tan^2 \theta}$$

### Half Angle Formulas
$$\sin^2 \theta = \frac{1 - \cos(2\theta)}{2}$$
$$\cos^2 \theta = \frac{1 + \cos(2\theta)}{2}$$

These are used extensively in integration!

```python
# Verify identities numerically
print("=" * 70)
print("VERIFYING TRIGONOMETRIC IDENTITIES")
print("=" * 70)

# Test angle
theta = np.pi / 5  # arbitrary angle

print(f"\nUsing θ = π/5 ≈ {theta:.4f} radians ≈ {np.degrees(theta):.2f}°")
print("\n" + "-" * 70)

# Pythagorean Identity
print("\n1. PYTHAGOREAN IDENTITY: sin²θ + cos²θ = 1")
result1 = np.sin(theta)**2 + np.cos(theta)**2
print(f"   sin²({theta:.4f}) + cos²({theta:.4f}) = {result1:.10f}")
print(f"   Expected: 1.0000000000")
print(f"   ✓ Identity verified!" if np.isclose(result1, 1.0) else "   ✗ Failed")

# Tan/Sec Identity
print("\n2. PYTHAGOREAN IDENTITY: 1 + tan²θ = sec²θ")
left2 = 1 + np.tan(theta)**2
right2 = (1/np.cos(theta))**2
print(f"   Left side:  1 + tan²θ = {left2:.10f}")
print(f"   Right side: sec²θ = {right2:.10f}")
print(f"   ✓ Identity verified!" if np.isclose(left2, right2) else "   ✗ Failed")

# Double angle formula for sine
print("\n3. DOUBLE ANGLE: sin(2θ) = 2sin(θ)cos(θ)")
left3 = np.sin(2 * theta)
right3 = 2 * np.sin(theta) * np.cos(theta)
print(f"   Left side:  sin(2θ) = {left3:.10f}")
print(f"   Right side: 2sin(θ)cos(θ) = {right3:.10f}")
print(f"   ✓ Identity verified!" if np.isclose(left3, right3) else "   ✗ Failed")

# Double angle formula for cosine
print("\n4. DOUBLE ANGLE: cos(2θ) = cos²θ - sin²θ")
left4 = np.cos(2 * theta)
right4 = np.cos(theta)**2 - np.sin(theta)**2
print(f"   Left side:  cos(2θ) = {left4:.10f}")
print(f"   Right side: cos²θ - sin²θ = {right4:.10f}")
print(f"   ✓ Identity verified!" if np.isclose(left4, right4) else "   ✗ Failed")

# Sum formula
alpha = np.pi / 4
beta = np.pi / 6
print(f"\n5. SUM FORMULA: sin(α + β) = sin(α)cos(β) + cos(α)sin(β)")
print(f"   Using α = π/4 and β = π/6")
left5 = np.sin(alpha + beta)
right5 = np.sin(alpha) * np.cos(beta) + np.cos(alpha) * np.sin(beta)
print(f"   Left side:  sin(π/4 + π/6) = {left5:.10f}")
print(f"   Right side: sin(π/4)cos(π/6) + cos(π/4)sin(π/6) = {right5:.10f}")
print(f"   ✓ Identity verified!" if np.isclose(left5, right5) else "   ✗ Failed")

print("\n" + "=" * 70)
print("\nAll identities verified! These are mathematical truths.")
print("Memorize them - you'll use them constantly in calculus!")
print("=" * 70)
```

## 5. Graphs of Trigonometric Functions

Understanding the graphs is essential for calculus!

### Key Properties

**sin(x) and cos(x)**:
- Period: $2\pi$

## 6. Inverse Trigonometric Functions

Inverse trig functions answer: "What angle has this trig value?"

### Definitions and Ranges

| Function | Definition | Domain | Range |
|----------|------------|--------|-------|
| $\arcsin(x)$ or $\sin^{-1}(x)$ | $y$ such that $\sin(y) = x$ | $[-1, 1]$ | $[-\pi/2, \pi/2]$ |

```python
# Graph inverse trig functions
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# arcsin(x)
x1 = np.linspace(-1, 1, 1000)
y1 = np.arcsin(x1)
axes[0].plot(x1, y1, 'b-', linewidth=3, label='y = arcsin(x)')
axes[0].axhline(y=0, color='k', linewidth=0.5)
axes[0].axvline(x=0, color='k', linewidth=0.5)
axes[0].axhline(y=np.pi/2, color='r', linestyle='--', alpha=0.5)
axes[0].axhline(y=-np.pi/2, color='r', linestyle='--', alpha=0.5)
axes[0].set_title('arcsin(x): Domain [-1,1], Range [-π/2, π/2]', fontsize=13, fontweight='bold')
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('y (radians)', fontsize=12)
axes[0].grid(True, alpha=0.3)
axes[0].legend(fontsize=11)
axes[0].set_yticks([-np.pi/2, -np.pi/4, 0, np.pi/4, np.pi/2])
axes[0].set_yticklabels(['-π/2', '-π/4', '0', 'π/4', 'π/2'])
axes[0].set_ylim(-2, 2)

# arccos(x)
y2 = np.arccos(x1)
axes[1].plot(x1, y2, 'r-', linewidth=3, label='y = arccos(x)')
axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].axvline(x=0, color='k', linewidth=0.5)
axes[1].axhline(y=np.pi, color='b', linestyle='--', alpha=0.5)
axes[1].axhline(y=0, color='b', linestyle='--', alpha=0.5)
axes[1].set_title('arccos(x): Domain [-1,1], Range [0, π]', fontsize=13, fontweight='bold')
axes[1].set_xlabel('x', fontsize=12)
axes[1].set_ylabel('y (radians)', fontsize=12)
axes[1].grid(True, alpha=0.3)
axes[1].legend(fontsize=11)
axes[1].set_yticks([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi])
axes[1].set_yticklabels(['0', 'π/4', 'π/2', '3π/4', 'π'])
axes[1].set_ylim(-0.5, 3.5)

# arctan(x)
x3 = np.linspace(-10, 10, 1000)
y3 = np.arctan(x3)
axes[2].plot(x3, y3, 'g-', linewidth=3, label='y = arctan(x)')
axes[2].axhline(y=0, color='k', linewidth=0.5)
axes[2].axvline(x=0, color='k', linewidth=0.5)
axes[2].axhline(y=np.pi/2, color='purple', linestyle='--', alpha=0.5, label='y = π/2 (asymptote)')
axes[2].axhline(y=-np.pi/2, color='purple', linestyle='--', alpha=0.5, label='y = -π/2 (asymptote)')
axes[2].set_title('arctan(x): Domain (-∞,∞), Range (-π/2, π/2)', fontsize=13, fontweight='bold')
axes[2].set_xlabel('x', fontsize=12)
axes[2].set_ylabel('y (radians)', fontsize=12)
axes[2].grid(True, alpha=0.3)
axes[2].legend(fontsize=11)
axes[2].set_yticks([-np.pi/2, -np.pi/4, 0, np.pi/4, np.pi/2])
axes[2].set_yticklabels(['-π/2', '-π/4', '0', 'π/4', 'π/2'])
axes[2].set_ylim(-2, 2)

plt.tight_layout()
plt.show()

print("Key Points:")
print("  • Inverse functions 'undo' the original function")
print("  • Ranges are restricted to make them functions (pass vertical line test)")
print("  • arcsin and arctan are odd functions: f(-x) = -f(x)")
print("  • arctan has horizontal asymptotes at ±π/2")
```

## 7. Trigonometry in Calculus: Derivatives

### Derivatives of Trig Functions (MEMORIZE!)

$$\frac{d}{dx}\sin(x) = \cos(x)$$
$$\frac{d}{dx}\cos(x) = -\sin(x)$$
$$\frac{d}{dx}\tan(x) = \sec^2(x)$$
$$\frac{d}{dx}\csc(x) = -\csc(x)\cot(x)$$
$$\frac{d}{dx}\sec(x) = \sec(x)\tan(x)$$
$$\frac{d}{dx}\cot(x) = -\csc^2(x)$$

### Why $\frac{d}{dx}\sin(x) = \cos(x)$?

From the limit definition:
$$\frac{d}{dx}\sin(x) = \lim_{h \to 0} \frac{\sin(x+h) - \sin(x)}{h}$$

Using the sum formula and the special limit $\lim_{h \to 0} \frac{\sin h}{h} = 1$:
$$= \lim_{h \to 0} \frac{\sin(x)\cos(h) + \cos(x)\sin(h) - \sin(x)}{h}$$
$$= \cos(x)$$

```python
# Visualize sin(x) and its derivative cos(x)
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

# sin and its derivative
axes[0, 0].plot(x, np.sin(x), 'b-', linewidth=3, label='f(x) = sin(x)')
axes[0, 0].plot(x, np.cos(x), 'r--', linewidth=2.5, label="f'(x) = cos(x)")
axes[0, 0].axhline(y=0, color='k', linewidth=0.5)
axes[0, 0].axvline(x=0, color='k', linewidth=0.5)
axes[0, 0].set_title('sin(x) and its derivative', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('x', fontsize=11)
axes[0, 0].set_ylabel('y', fontsize=11)
axes[0, 0].legend(fontsize=11)
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi])
axes[0, 0].set_xticklabels(['-2π', '-π', '0', 'π', '2π'])
axes[0, 0].text(np.pi/2, -0.5, 'sin has max\ncos = 0', fontsize=10,
               bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.5))

# cos and its derivative
axes[0, 1].plot(x, np.cos(x), 'b-', linewidth=3, label='f(x) = cos(x)')
axes[0, 1].plot(x, -np.sin(x), 'r--', linewidth=2.5, label="f'(x) = -sin(x)")
axes[0, 1].axhline(y=0, color='k', linewidth=0.5)
axes[0, 1].axvline(x=0, color='k', linewidth=0.5)
axes[0, 1].set_title('cos(x) and its derivative', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('x', fontsize=11)
axes[0, 1].set_ylabel('y', fontsize=11)
axes[0, 1].legend(fontsize=11)
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi])
axes[0, 1].set_xticklabels(['-2π', '-π', '0', 'π', '2π'])

# tan and its derivative
x_tan = x[np.abs(np.cos(x)) > 0.05]
axes[1, 0].plot(x_tan, np.tan(x_tan), 'b-', linewidth=2.5, label='f(x) = tan(x)')
axes[1, 0].plot(x_tan, 1/np.cos(x_tan)**2, 'r--', linewidth=2, label="f'(x) = sec²(x)")
for asymptote in [-3*np.pi/2, -np.pi/2, np.pi/2, 3*np.pi/2]:
    axes[1, 0].axvline(x=asymptote, color='gray', linestyle='--', alpha=0.3)
axes[1, 0].axhline(y=0, color='k', linewidth=0.5)
axes[1, 0].set_title('tan(x) and its derivative', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('x', fontsize=11)
axes[1, 0].set_ylabel('y', fontsize=11)
axes[1, 0].legend(fontsize=11)
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi])
axes[1, 0].set_xticklabels(['-2π', '-π', '0', 'π', '2π'])
axes[1, 0].set_ylim(-5, 10)

# Important limit: sin(x)/x → 1 as x → 0
x_limit = np.linspace(-2*np.pi, 2*np.pi, 1000)
x_limit_nonzero = x_limit[np.abs(x_limit) > 0.01]
y_limit = np.sin(x_limit_nonzero) / x_limit_nonzero
axes[1, 1].plot(x_limit_nonzero, y_limit, 'purple', linewidth=3, label='y = sin(x)/x')
axes[1, 1].plot(0, 1, 'ro', markersize=15, label='Limit = 1 at x=0')
axes[1, 1].axhline(y=1, color='r', linestyle='--', alpha=0.5)
axes[1, 1].axhline(y=0, color='k', linewidth=0.5)
axes[1, 1].axvline(x=0, color='k', linewidth=0.5)
axes[1, 1].set_title('Key Limit: lim(x→0) sin(x)/x = 1', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('x', fontsize=11)
axes[1, 1].set_ylabel('y', fontsize=11)
axes[1, 1].legend(fontsize=11)
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].set_xlim(-10, 10)
axes[1, 1].set_ylim(-0.5, 1.5)
axes[1, 1].text(5, 0.5, 'This limit is crucial\nfor proving d/dx(sin x) = cos x!', 
               fontsize=11, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.6))

plt.tight_layout()
plt.show()

print("Observation: Notice how the derivative tells us about the original function!")
print("  • Where sin(x) has slope 0 (max/min), cos(x) = 0")
print("  • Where sin(x) is increasing (positive slope), cos(x) > 0")
print("  • Where sin(x) is decreasing (negative slope), cos(x) < 0")
```

### Problem 1: Simplify using the Pythagorean identity

Simplify: $\sin^2(x) + \cos^2(x) + \tan^2(x)$

```python
print("Problem 1: Simplify sin²(x) + cos²(x) + tan²(x)")
print("\nSolution:")
print("  Step 1: Use Pythagorean identity: sin²(x) + cos²(x) = 1")
print("  Step 2: Substitute: 1 + tan²(x)")
print("  Step 3: Use identity: 1 + tan²(x) = sec²(x)")
print("\nAnswer: sec²(x)")

# Verify numerically
x_test = np.pi / 7
original = np.sin(x_test)**2 + np.cos(x_test)**2 + np.tan(x_test)**2
simplified = 1 / np.cos(x_test)**2
print(f"\nNumerical verification at x = π/7:")
print(f"  Original expression: {original:.10f}")
print(f"  Simplified (sec²x):  {simplified:.10f}")
print(f"  Match: {np.isclose(original, simplified)}")
```

### Problem 2: Evaluate using special angles

Find the exact value: $\sin(\pi/6) + \cos(\pi/3) - \tan(\pi/4)$

```python
print("Problem 2: sin(π/6) + cos(π/3) - tan(π/4)")
print("\nSolution:")
print("  From the unit circle:")
print("    sin(π/6) = sin(30°) = 1/2")
print("    cos(π/3) = cos(60°) = 1/2")
print("    tan(π/4) = tan(45°) = 1")
print("\n  Therefore:")
print("    1/2 + 1/2 - 1 = 1 - 1 = 0")
print("\nAnswer: 0")

# Verify
result = np.sin(np.pi/6) + np.cos(np.pi/3) - np.tan(np.pi/4)
print(f"\nNumerical check: {result:.10f}")
```

### Problem 3: Use a trig identity

Prove that: $\frac{\sin(2x)}{\sin(x)} = 2\cos(x)$

```python
print("Problem 3: Prove that sin(2x)/sin(x) = 2cos(x)")
print("\nSolution:")
print("  Step 1: Use double angle formula: sin(2x) = 2sin(x)cos(x)")
print("  Step 2: Substitute into left side:")
print("          sin(2x)/sin(x) = [2sin(x)cos(x)]/sin(x)")
print("  Step 3: Cancel sin(x) (assuming sin(x) ≠ 0):")
print("          = 2cos(x)")
print("\n  Therefore: sin(2x)/sin(x) = 2cos(x) ✓")

# Verify for multiple values
print("\nNumerical verification:")
test_angles = [np.pi/6, np.pi/4, np.pi/3, 2*np.pi/3]
for angle in test_angles:
    left = np.sin(2*angle) / np.sin(angle)
    right = 2 * np.cos(angle)
    print(f"  x = {angle:.4f}: LHS = {left:.6f}, RHS = {right:.6f}, Match = {np.isclose(left, right)}")
```

### Problem 4: Find the derivative

Find $\frac{d}{dx}[x^2 \sin(x)]$

```python
print("Problem 4: Find d/dx[x²·sin(x)]")
print("\nSolution:")
print("  Use the Product Rule: d/dx[f·g] = f'·g + f·g'")
print("\n  Let f = x²,  f' = 2x")
print("  Let g = sin(x),  g' = cos(x)")
print("\n  Therefore:")
print("    d/dx[x²·sin(x)] = (2x)·sin(x) + x²·cos(x)")
print("                    = 2x·sin(x) + x²·cos(x)")
print("\nAnswer: 2x·sin(x) + x²·cos(x)")

# Verify numerically using the definition of derivative
x_test = 1.5
h = 0.0001
f = lambda x: x**2 * np.sin(x)
numerical_derivative = (f(x_test + h) - f(x_test)) / h
formula_derivative = 2*x_test*np.sin(x_test) + x_test**2*np.cos(x_test)

print(f"\nNumerical verification at x = {x_test}:")
print(f"  From limit definition: {numerical_derivative:.8f}")
print(f"  From formula:          {formula_derivative:.8f}")
print(f"  Match: {np.isclose(numerical_derivative, formula_derivative)}")
```

## 9. Key Takeaways

### Must Memorize

**Unit Circle Values** (30°, 45°, 60°):
- sin(30°) = 1/2, cos(30°) = √3/2
- sin(45°) = √2/2, cos(45°) = √2/2
- sin(60°) = √3/2, cos(60°) = 1/2

**Key Identities**:
- sin²θ + cos²θ = 1
- 1 + tan²θ = sec²θ
- sin(2θ) = 2sin(θ)cos(θ)
- cos(2θ) = cos²θ - sin²θ

**Derivatives**:
- d/dx(sin x) = cos x
- d/dx(cos x) = -sin x
- d/dx(tan x) = sec²x
