# Integration: Accumulation and Area

## What is Integration?

Integration has **two related but different interpretations**:

### 1. Indefinite Integral (Antiderivative)
**Find a function whose derivative is the given function**

$$\int f(x)\,dx = F(x) + C \quad \text{where } F'(x) = f(x)$$

### 2. Definite Integral (Area/Accumulation)
**Find the total accumulation or signed area under a curve**

$$\int_a^b f(x)\,dx = \text{signed area between } f(x) \text{ and x-axis from } a \text{ to } b$$

## The Fundamental Theorem of Calculus

This theorem connects derivatives and integrals!

**Part 1**: If $F(x) = \int_a^x f(t)\,dt$, then $F'(x) = f(x)$

**Part 2**: If $F'(x) = f(x)$, then $\int_a^b f(x)\,dx = F(b) - F(a)$

**Meaning**: Differentiation and integration are inverse operations!

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import integrate

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12
```

## 1. Understanding Area and Riemann Sums

The definite integral is defined as the limit of Riemann sums:

$$\int_a^b f(x)\,dx = \lim_{n \to \infty} \sum_{i=1}^n f(x_i^*) \Delta x$$

where $\Delta x = \frac{b-a}{n}$ and $x_i^*$ is a sample point in the $i$-th subinterval.

**Three common choices**:
- **Left Riemann Sum**: $x_i^* = x_{i-1}$ (left endpoint)
- **Right Riemann Sum**: $x_i^* = x_i$ (right endpoint)
- **Midpoint Rule**: $x_i^* = \frac{x_{i-1} + x_i}{2}$ (midpoint)

```python
# Visualize Riemann sums
def f(x):
    return 0.5*x**2 + 1

a, b = 0, 4
x = np.linspace(a, b, 1000)
y = f(x)

# True integral value
true_integral = (0.5/3) * (b**3 - a**3) + (b - a)  # ∫(0.5x² + 1)dx from 0 to 4

fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# Different numbers of rectangles
n_values = [4, 8, 16, 32, 64, 128]

for idx, (ax, n) in enumerate(zip(axes.flat, n_values)):
    ax.plot(x, y, 'b-', linewidth=2, label='f(x) = 0.5x² + 1')
    ax.fill_between(x, 0, y, alpha=0.2, color='blue')
    
    # Compute Riemann sum (using left endpoints)
    dx = (b - a) / n
    x_left = np.linspace(a, b - dx, n)
    riemann_sum = 0
    
    for xi in x_left:
        height = f(xi)
        riemann_sum += height * dx
        rect = plt.Rectangle((xi, 0), dx, height, 
                            edgecolor='red', facecolor='red', 
                            alpha=0.3, linewidth=1.5)
        ax.add_patch(rect)
    
    error = abs(riemann_sum - true_integral)
    
    ax.set_xlim(a - 0.2, b + 0.2)
    ax.set_ylim(0, max(y) * 1.1)
    ax.set_title(f'n = {n} rectangles\nSum ≈ {riemann_sum:.4f}, Error = {error:.4f}', 
                 fontsize=12, fontweight='bold')
    ax.set_xlabel('x', fontsize=11)
    ax.set_ylabel('f(x)', fontsize=11)
    ax.grid(True, alpha=0.3)
    
plt.suptitle(f'Riemann Sums Approaching True Integral = {true_integral:.4f}', 
             fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("As n → ∞, the Riemann sum approaches the true integral!")
print(f"\nTrue integral value: {true_integral:.6f}")
print("\nRiemann sum approximations:")
for n in [4, 8, 16, 32, 64, 128, 256, 512, 1024]:
    dx = (b - a) / n
    x_left = np.linspace(a, b - dx, n)
    riemann_sum = sum(f(xi) * dx for xi in x_left)
    error = abs(riemann_sum - true_integral)
    print(f"  n = {n:4d}  →  Sum = {riemann_sum:.6f}  (error: {error:.6f})")
```

## 2. Basic Integration Rules

### Power Rule (Reverse of Derivative Power Rule)
$$\int x^n\,dx = \frac{x^{n+1}}{n+1} + C \quad (n \neq -1)$$

### Special Case: $n = -1$
$$\int \frac{1}{x}\,dx = \ln|x| + C$$

### Constant Multiple
$$\int cf(x)\,dx = c\int f(x)\,dx$$

### Sum/Difference
$$\int [f(x) \pm g(x)]\,dx = \int f(x)\,dx \pm \int g(x)\,dx$$

### Common Integrals
$$\int e^x\,dx = e^x + C$$
$$\int a^x\,dx = \frac{a^x}{\ln a} + C$$
$$\int \sin x\,dx = -\cos x + C$$
$$\int \cos x\,dx = \sin x + C$$
$$\int \sec^2 x\,dx = \tan x + C$$
$$\int \frac{1}{x^2+1}\,dx = \arctan x + C$$
$$\int \frac{1}{\sqrt{1-x^2}}\,dx = \arcsin x + C$$

```python
print("=" * 70)
print("BASIC INTEGRATION EXAMPLES")
print("=" * 70)

print("\n1. POWER RULE: ∫x^n dx = x^(n+1)/(n+1) + C")
print("-" * 70)
examples = [
    ("∫x² dx", "x³/3 + C"),
    ("∫x⁴ dx", "x⁵/5 + C"),
    ("∫√x dx = ∫x^(1/2) dx", "(2/3)x^(3/2) + C"),
    ("∫1/x² dx = ∫x^(-2) dx", "-1/x + C"),
]
for problem, solution in examples:
    print(f"  {problem:30} = {solution}")

print("\n2. SUM/DIFFERENCE RULE")
print("-" * 70)
print("  ∫(3x² + 2x - 5) dx = ∫3x² dx + ∫2x dx - ∫5 dx")
print("                     = 3(x³/3) + 2(x²/2) - 5x + C")
print("                     = x³ + x² - 5x + C")

print("\n3. EXPONENTIAL AND LOGARITHMIC")
print("-" * 70)
examples = [
    ("∫eˣ dx", "eˣ + C"),
    ("∫2ˣ dx", "2ˣ/ln(2) + C"),
    ("∫1/x dx", "ln|x| + C"),
]
for problem, solution in examples:
    print(f"  {problem:25} = {solution}")

print("\n4. TRIGONOMETRIC")
print("-" * 70)
examples = [
    ("∫sin(x) dx", "-cos(x) + C"),
    ("∫cos(x) dx", "sin(x) + C"),
    ("∫sec²(x) dx", "tan(x) + C"),
]
for problem, solution in examples:
    print(f"  {problem:25} = {solution}")

print("\n5. VERIFYING BY DIFFERENTIATION")
print("-" * 70)
print("  If ∫f(x) dx = F(x) + C, then d/dx[F(x)] should equal f(x)")
print("")
print("  Example: ∫x² dx = x³/3 + C")
print("  Check: d/dx[x³/3 + C] = (1/3)·3x² = x² ✓")

print("\n" + "=" * 70)
```

## 3. The Fundamental Theorem of Calculus (FTC)

### Part 1 (Derivative of an Integral)
If $F(x) = \int_a^x f(t)\,dt$, then $F'(x) = f(x)$

### Part 2 (Evaluation Theorem)
If $F'(x) = f(x)$, then:
$$\int_a^b f(x)\,dx = F(b) - F(a) = F(x)\Big|_a^b$$

```python
# Visualize FTC Part 1
def f(x):
    return np.sin(x) + 1

x = np.linspace(0, 2*np.pi, 1000)
a = 0

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Left: Show f(x) and accumulation function F(x)
ax1.plot(x, f(x), 'b-', linewidth=3, label='f(x) = sin(x) + 1')
ax1.fill_between(x[:500], 0, f(x[:500]), alpha=0.3, color='blue', 
                 label='Area = F(x₀)')
ax1.axvline(x=x[499], color='r', linestyle='--', linewidth=2, label='x = x₀')
ax1.set_xlabel('x', fontsize=13)
ax1.set_ylabel('f(x)', fontsize=13)
ax1.set_title('f(x): The Integrand', fontsize=15, fontweight='bold')
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.set_ylim(-0.5, 2.5)

# Right: Show F(x) = integral from a to x
F = np.array([integrate.quad(f, a, xi)[0] for xi in x])
ax2.plot(x, F, 'r-', linewidth=3, label='F(x) = ∫₀ˣ f(t) dt')
ax2.set_xlabel('x', fontsize=13)
ax2.set_ylabel('F(x)', fontsize=13)
ax2.set_title('F(x): The Accumulation Function\nNote: F\'(x) = f(x)', 
             fontsize=15, fontweight='bold')
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("FTC Part 1: The derivative of the accumulation function equals the original function!")
print("  If F(x) = ∫₀ˣ (sin(t) + 1) dt")
print("  Then F'(x) = sin(x) + 1 = f(x)")

# Verify numerically
print("\nNumerical verification:")
x_test = np.pi
h = 0.0001
F_test = integrate.quad(f, a, x_test)[0]
F_test_h = integrate.quad(f, a, x_test + h)[0]
derivative_numerical = (F_test_h - F_test) / h
derivative_actual = f(x_test)
print(f"  At x = π:")
print(f"    F'(π) ≈ {derivative_numerical:.6f}  (numerical derivative)")
print(f"    f(π) = {derivative_actual:.6f}  (actual function value)")
print(f"    Match! ✓")
```

```python
# Demonstrate FTC Part 2
print("=" * 70)
print("FUNDAMENTAL THEOREM OF CALCULUS - PART 2")
print("=" * 70)

print("\nEXAMPLE 1: Evaluate ∫₁⁴ x² dx")
print("-" * 70)
print("Step 1: Find antiderivative")
print("        ∫x² dx = x³/3 + C")
print("")
print("Step 2: Apply FTC Part 2")
print("        ∫₁⁴ x² dx = [x³/3]₁⁴")
print("                  = (4³/3) - (1³/3)")
print("                  = 64/3 - 1/3")
print("                  = 63/3")
print("                  = 21")

# Verify
result = (4**3 / 3) - (1**3 / 3)
print(f"\nResult: {result}")

# Visualize
x = np.linspace(0, 5, 1000)
y = x**2
x_region = x[(x >= 1) & (x <= 4)]
y_region = x_region**2

plt.figure(figsize=(12, 7))
plt.plot(x, y, 'b-', linewidth=3, label='f(x) = x²')
plt.fill_between(x_region, 0, y_region, alpha=0.4, color='blue', 
                 label=f'Area = ∫₁⁴ x² dx = {result:.0f}')
plt.axvline(x=1, color='r', linestyle='--', linewidth=2, alpha=0.7, label='x = 1')
plt.axvline(x=4, color='r', linestyle='--', linewidth=2, alpha=0.7, label='x = 4')
plt.xlabel('x', fontsize=13)
plt.ylabel('f(x)', fontsize=13)
plt.title('Definite Integral = Signed Area', fontsize=15, fontweight='bold')
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.xlim(0, 5)
plt.ylim(0, 20)
plt.show()

print("\n" + "=" * 70)
print("\nEXAMPLE 2: Evaluate ∫₀^π sin(x) dx")
print("-" * 70)
print("Step 1: Find antiderivative")
print("        ∫sin(x) dx = -cos(x) + C")
print("")
print("Step 2: Apply FTC Part 2")
print("        ∫₀^π sin(x) dx = [-cos(x)]₀^π")
print("                       = -cos(π) - (-cos(0))")
print("                       = -(-1) - (-1)")
print("                       = 1 + 1")
print("                       = 2")

result2 = -np.cos(np.pi) - (-np.cos(0))
print(f"\nResult: {result2:.6f}")

print("\n" + "=" * 70)
```

## 4. Integration Techniques

### U-Substitution (Chain Rule in Reverse)

**Pattern**: $\int f(g(x))g'(x)\,dx$

**Method**:
1. Choose $u = g(x)$
2. Compute $du = g'(x)dx$
3. Substitute: $\int f(u)\,du$
4. Integrate
5. Substitute back

**Example**: $\int 2x\cos(x^2)\,dx$
- Let $u = x^2$, then $du = 2x\,dx$
- $\int \cos(u)\,du = \sin(u) + C = \sin(x^2) + C$

```python
print("=" * 70)
print("U-SUBSTITUTION EXAMPLES")
print("=" * 70)

print("\nEXAMPLE 1: ∫2x·e^(x²) dx")
print("-" * 70)
print("Step 1: Identify u")
print("        Notice: derivative of x² is 2x (which we have!)")
print("        Let u = x²")
print("")
print("Step 2: Find du")
print("        du = 2x dx")
print("")
print("Step 3: Substitute")
print("        ∫2x·e^(x²) dx = ∫e^u du")
print("")
print("Step 4: Integrate")
print("        = e^u + C")
print("")
print("Step 5: Substitute back")
print("        = e^(x²) + C")
print("")
print("Verify: d/dx[e^(x²)] = e^(x²)·2x ✓")

print("\n" + "=" * 70)
print("\nEXAMPLE 2: ∫x√(x²+1) dx")
print("-" * 70)
print("Step 1: Let u = x² + 1")
print("        (derivative of x²+1 is 2x, we have x)")
print("")
print("Step 2: du = 2x dx  →  x dx = du/2")
print("")
print("Step 3: Substitute")
print("        ∫x√(x²+1) dx = ∫√u · (du/2)")
print("                     = (1/2)∫u^(1/2) du")
print("")
print("Step 4: Integrate")
print("        = (1/2) · (u^(3/2))/(3/2) + C")
print("        = (1/3)u^(3/2) + C")
print("")
print("Step 5: Substitute back")
print("        = (1/3)(x²+1)^(3/2) + C")

print("\n" + "=" * 70)
print("\nEXAMPLE 3: ∫sin(x)cos(x) dx")
print("-" * 70)
print("Method 1: Let u = sin(x), du = cos(x) dx")
print("          ∫sin(x)cos(x) dx = ∫u du = u²/2 + C = sin²(x)/2 + C")
print("")
print("Method 2: Let u = cos(x), du = -sin(x) dx")
print("          ∫sin(x)cos(x) dx = -∫u du = -u²/2 + C = -cos²(x)/2 + C")
print("")
print("Both correct! (They differ by a constant using sin²x + cos²x = 1)")

print("\n" + "=" * 70)
```

### Integration by Parts (Product Rule in Reverse)

$$\int u\,dv = uv - \int v\,du$$

**Strategy**: Choose $u$ and $dv$ so that $\int v\,du$ is simpler than $\int u\,dv$

**LIATE Rule** (priority for choosing $u$):
1. **L**ogarithmic: $\ln x, \log x$
2. **I**nverse trig: $\arcsin x, \arctan x$
3. **A**lgebraic: $x, x^2, x^3$
4. **T**rigonometric: $\sin x, \cos x$
5. **E**xponential: $e^x, a^x$

```python
print("=" * 70)
print("INTEGRATION BY PARTS EXAMPLES")
print("=" * 70)

print("\nFormula: ∫u dv = uv - ∫v du")
print("")

print("\nEXAMPLE 1: ∫x·eˣ dx")
print("-" * 70)
print("Step 1: Choose u and dv using LIATE")
print("        u = x       (Algebraic)")
print("        dv = eˣ dx  (Exponential)")
print("")
print("Step 2: Find du and v")
print("        du = dx")
print("        v = ∫eˣ dx = eˣ")
print("")
print("Step 3: Apply formula")
print("        ∫x·eˣ dx = x·eˣ - ∫eˣ dx")
print("                 = x·eˣ - eˣ + C")
print("                 = eˣ(x - 1) + C")
print("")
print("Verify: d/dx[eˣ(x-1)] = eˣ(x-1) + eˣ = eˣ·x ✓")

print("\n" + "=" * 70)
print("\nEXAMPLE 2: ∫ln(x) dx")
print("-" * 70)
print("Step 1: Choose u and dv")
print("        u = ln(x)   (Logarithmic - highest priority!)")
print("        dv = dx")
print("")
print("Step 2: Find du and v")
print("        du = (1/x) dx")
print("        v = x")
print("")
print("Step 3: Apply formula")
print("        ∫ln(x) dx = x·ln(x) - ∫x·(1/x) dx")
print("                  = x·ln(x) - ∫1 dx")
print("                  = x·ln(x) - x + C")
print("                  = x(ln(x) - 1) + C")

print("\n" + "=" * 70)
print("\nEXAMPLE 3: ∫x²·sin(x) dx  (Need to use parts twice!)")
print("-" * 70)
print("First application:")
print("  u = x²,  dv = sin(x) dx")
print("  du = 2x dx,  v = -cos(x)")
print("  ∫x²sin(x) dx = -x²cos(x) + ∫2x·cos(x) dx")
print("")
print("Second application on ∫2x·cos(x) dx:")
print("  u = 2x,  dv = cos(x) dx")
print("  du = 2 dx,  v = sin(x)")
print("  ∫2x·cos(x) dx = 2x·sin(x) - ∫2sin(x) dx")
print("                = 2x·sin(x) + 2cos(x) + C")
print("")
print("Combine:")
print("  ∫x²sin(x) dx = -x²cos(x) + 2x·sin(x) + 2cos(x) + C")
print("               = (2-x²)cos(x) + 2x·sin(x) + C")

print("\n" + "=" * 70)
```

### Partial Fractions

For rational functions $\frac{P(x)}{Q(x)}$ where degree of $P <$ degree of $Q$:

1. Factor denominator $Q(x)$
2. Decompose into partial fractions
3. Integrate each term separately

**Example**: $\int \frac{1}{x^2-1}\,dx$

Factor: $x^2 - 1 = (x-1)(x+1)$

Decompose: $\frac{1}{(x-1)(x+1)} = \frac{A}{x-1} + \frac{B}{x+1}$

Solve: $A = \frac{1}{2}, B = -\frac{1}{2}$

Integrate: $\int \left(\frac{1/2}{x-1} - \frac{1/2}{x+1}\right)dx = \frac{1}{2}\ln|x-1| - \frac{1}{2}\ln|x+1| + C$

## 5. Applications of Integration

### Area Between Curves

Area between $f(x)$ and $g(x)$ from $a$ to $b$:
$$A = \int_a^b |f(x) - g(x)|\,dx$$

If $f(x) \geq g(x)$ on $[a,b]$:
$$A = \int_a^b [f(x) - g(x)]\,dx$$

### Volume of Revolution

**Disk Method** (revolving around x-axis):
$$V = \pi \int_a^b [f(x)]^2\,dx$$

**Washer Method** (between two curves):
$$V = \pi \int_a^b [R(x)]^2 - [r(x)]^2\,dx$$

where $R(x)$ is outer radius, $r(x)$ is inner radius

```python
# Visualize area between curves
x = np.linspace(0, np.pi, 1000)
f = np.sin(x) + 1
g = np.cos(x)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Area between curves
ax1.plot(x, f, 'b-', linewidth=3, label='f(x) = sin(x) + 1')
ax1.plot(x, g, 'r-', linewidth=3, label='g(x) = cos(x)')
ax1.fill_between(x, g, f, where=(f >= g), alpha=0.4, color='green', 
                 label='Area between curves')
ax1.set_xlabel('x', fontsize=13)
ax1.set_ylabel('y', fontsize=13)
ax1.set_title('Area Between Curves\nA = ∫[f(x) - g(x)] dx', fontsize=14, fontweight='bold')
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, np.pi)

# Compute the area
area = integrate.quad(lambda x: np.sin(x) + 1 - np.cos(x), 0, np.pi)[0]
ax1.text(np.pi/2, 1, f'Area ≈ {area:.3f}', fontsize=13, ha='center',
         bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Volume of revolution
x2 = np.linspace(0, 2, 100)
y2 = np.sqrt(x2)

ax2.plot(x2, y2, 'b-', linewidth=3, label='y = √x')
ax2.fill_between(x2, 0, y2, alpha=0.3, color='blue')
ax2.set_xlabel('x', fontsize=13)
ax2.set_ylabel('y', fontsize=13)
ax2.set_title('Revolution around x-axis\nV = π∫[f(x)]² dx', fontsize=14, fontweight='bold')
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.set_aspect('equal')

# Compute volume
volume = np.pi * integrate.quad(lambda x: x, 0, 2)[0]  # ∫x dx since (√x)² = x
ax2.text(1, 0.5, f'Volume = π∫₀² x dx\n= π[x²/2]₀²\n= 2π ≈ {volume:.3f}', 
         fontsize=12, ha='center',
         bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

plt.tight_layout()
plt.show()
```

4. **For definite integrals**: Use FTC Part 2
